```python
import numpy as np

# Create a list of numbers
data = np.array([10, 20, 20, 30, 40, 50])

# Calculate mean
mean = np.mean(data)

# Calculate median
median = np.median(data)

# Calculate mode
values, counts = np.unique(data, return_counts=True)
mode = values[np.argmax(counts)]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```
