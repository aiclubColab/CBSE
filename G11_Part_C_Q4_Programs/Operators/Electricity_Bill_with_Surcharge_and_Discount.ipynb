Task:
Write a Python program that:

*   Accepts the number of electricity units consumed.
*   Calculates the electricity bill using different rates for different unit ranges.
*   Adds a 5% surcharge if the bill exceeds ₹2000.
*   Gives a 10% discount if the consumption is below 150 units.
*   Displays the final electricity bill.

```python
#Asks the user to enter the electricity units consumed and converts the input into an integer.
units = int(input("Enter units consumed: "))
```

```python
#check if the consumption is 100 units or less and calculate the bill at ₹2 per unit.
if units <= 100:
    bill = units * 2

#Checks if the consumption is between 101 and 300 units and calculates ₹2 for the first 100 units and ₹3 for the remaining units.
elif units <= 300:
    bill = (100 * 2) + (units - 100) * 3

#Runs when the consumption is above 300 units and calculates ₹2 for the first 100, ₹3 for the next 200, and ₹5 for units above 300.
else:
    bill = (100 * 2) + (200 * 3) + (units - 300) * 5
```

```python
#Checks if the bill is more than ₹2000 and adds a 5% surcharge to the bill.
if bill > 2000:
    bill += bill * 5 / 100

#Checks if the consumption is below 150 units and gives a 10% discount on the bill.
elif units < 150:
    bill -= bill * 10 / 100
```

```python
#Displays the final electricity bill.
print("Final Electricity Bill: ₹", bill)
```
