```python
#Write a Python program to calculate an electricity bill based on units consumed.
#First 100 units: ₹2/unit
#Next 200 units: ₹3/unit
#Above 300 units: ₹5/unit
#If the bill exceeds ₹2000, add a 5% surcharge.
#If consumption is below 150 units, give a 10% discount.
#Display the final bill.

units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 2
elif units <= 300:
    bill = (100 * 2) + (units - 100) * 3
else:
    bill = (100 * 2) + (200 * 3) + (units - 300) * 5

if bill > 2000:
    bill += bill * 5 / 100
elif units < 150:
    bill -= bill * 10 / 100

print("Final Electricity Bill: ₹", bill)
```
