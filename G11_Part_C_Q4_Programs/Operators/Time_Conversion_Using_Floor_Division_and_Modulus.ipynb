**Task:**

To write a Python program that:
*   Accepts a total number of seconds from the user.
*   Converts the total seconds into:

                                    1. Days
                                    2. Hours
                                    3. Minutes
                                    4. Seconds
*   Displays each value separately.

```python
#Takes the total number of seconds as an integer.
total_seconds = int(input("Enter total seconds: "))
```

```python
#Calculates the number of complete days using 86,400 seconds per day.
days = total_seconds // 86400

#Finds the remaining seconds after calculating the days.
remaining = total_seconds % 86400
```

```python
#Calculates the number of complete hours using 3,600 seconds per hour.
hours = remaining // 3600

#Finds the remaining seconds after calculating the hours.
remaining = remaining % 3600
```

```python
#Calculates the number of complete minutes using 60 seconds per minute.
minutes = remaining // 60

#Finds the remaining seconds after calculating the minutes.
seconds = remaining % 60
```

```python
#Displays the number of days.
print("Days:", days)

#Displays the number of hours.
print("Hours:", hours)

#Displays the number of minutes.
print("Minutes:", minutes)

#Displays the number of seconds.
print("Seconds:", seconds)
```
