**Task:**

To write a Python program that accepts marks in five subjects and calculates:
*   Total marks
*   Average marks
*   Percentage
*   Whether the student has passed
*   Whether the student has achieved distinction
*   Whether the student is eligible for a scholarship

**Conditions**

*   A student must score at least 40 marks in every subject to pass.
*   A student must have an overall percentage of at least 75% to achieve distinction.
*   Scholarship eligibility requires:Percentage ≥ 90% and Marks in every subject ≥ 80.

```python
#Takes the marks for Subject 1 as a decimal number.
m1 = float(input("Enter marks in Subject 1: "))

#Takes the marks for Subject 2 as a decimal number.
m2 = float(input("Enter marks in Subject 2: "))

#Takes the marks for Subject 3 as a decimal number.
m3 = float(input("Enter marks in Subject 3: "))

#Takes the marks for Subject 4 as a decimal number.
m4 = float(input("Enter marks in Subject 4: "))

#Takes the marks for Subject 5 as a decimal number.
m5 = float(input("Enter marks in Subject 5: "))
```

```python
#Calculates the total marks of all five subjects.
total = m1 + m2 + m3 + m4 + m5

#Calculates the average marks.
average = total / 5

#Calculates the percentage based on a maximum of 500 marks.
percentage = (total / 500) * 100
```

```python
#Checks whether the student scored at least 40 marks in every subject.
passed = (m1 >= 40 and m2 >= 40 and
          m3 >= 40 and m4 >= 40 and m5 >= 40)
```

```python
#Checks whether the student passed and scored at least 75%.
distinction = passed and percentage >= 75
```

```python
#Checks whether the student scored 90% or more and at least 80 marks in every subject.
scholarship = (percentage >= 90 and
               m1 >= 80 and m2 >= 80 and
               m3 >= 80 and m4 >= 80 and m5 >= 80)
```

```python
#Displays the total marks.
print("\nTotal:", total)

#Displays the average marks.
print("Average:", average)

#Displays the percentage.
print("Percentage:", percentage)

#Displays whether the student passed (True) or failed (False).
print("Passed:", passed)

#Displays whether the student is eligible for distinction.
print("Distinction:", distinction)

#Displays whether the student is eligible for a scholarship.
print("Scholarship Eligible:", scholarship)
```
