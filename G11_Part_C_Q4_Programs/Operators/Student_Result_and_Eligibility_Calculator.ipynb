```python
'''
Task:
A student has marks in five subjects. Write a program to calculate:

Total marks
Average
Percentage
Whether the student has passed
Whether the student is eligible for a scholarship

Conditions:

Each subject must have at least 40 marks.
Overall percentage must be at least 75% to pass with distinction.
Scholarship eligibility requires percentage ≥ 90% and every subject ≥ 80.
'''

m1 = float(input("Enter marks in Subject 1: "))
m2 = float(input("Enter marks in Subject 2: "))
m3 = float(input("Enter marks in Subject 3: "))
m4 = float(input("Enter marks in Subject 4: "))
m5 = float(input("Enter marks in Subject 5: "))

total = m1 + m2 + m3 + m4 + m5
average = total / 5
percentage = (total / 500) * 100

passed = (m1 >= 40 and m2 >= 40 and
          m3 >= 40 and m4 >= 40 and m5 >= 40)

distinction = passed and percentage >= 75

scholarship = (percentage >= 90 and
               m1 >= 80 and m2 >= 80 and
               m3 >= 80 and m4 >= 80 and m5 >= 80)

print("\nTotal:", total)
print("Average:", average)
print("Percentage:", percentage)
print("Passed:", passed)
print("Distinction:", distinction)
print("Scholarship Eligible:", scholarship)
```
