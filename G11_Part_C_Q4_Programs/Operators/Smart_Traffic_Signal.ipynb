```python
'''
A traffic signal should allow a vehicle to move when:

The signal is green and there is no pedestrian crossing.

OR

The signal is yellow and the vehicle is emergency service.

But vehicles must stop if the emergency stop system is active.
'''

signal = input("Enter signal (red/yellow/green): ")
pedestrian = input("Is pedestrian crossing? (yes/no): ")
emergency = input("Is the vehicle an emergency vehicle? (yes/no): ")
emergency_stop = input("Is emergency stop active? (yes/no): ")

normal_move = (
    signal == "green" and
    pedestrian == "no"
)

emergency_move = (
    signal == "yellow" and
    emergency == "yes"
)

can_move = (
    (normal_move or emergency_move) and
    emergency_stop == "no"
)

print("Vehicle Can Move:", can_move)
```
