Task:
To write a Python program that determines whether a vehicle is allowed to move based on:
*   The traffic signal colour.
*   Whether a pedestrian is crossing.
*   Whether the vehicle is an emergency vehicle.
*   Whether the emergency stop system is active.

The vehicle can move when:
The signal is green and there is no pedestrian crossing
OR
The signal is yellow and the vehicle is an emergency vehicle.

However, the vehicle must stop if the emergency stop system is active, regardless of the other conditions.

```python
#Asks the user to enter the traffic signal colour.
signal = input("Enter signal (red/yellow/green): ")

#Asks whether a pedestrian is crossing.
pedestrian = input("Is pedestrian crossing? (yes/no): ")

#Asks whether the vehicle is an emergency vehicle.
emergency = input("Is the vehicle an emergency vehicle? (yes/no): ")

#Asks whether the emergency stop is active.
emergency_stop = input("Is emergency stop active? (yes/no): ")
```

```python
#Checks if the signal is green and no pedestrian is crossing, allowing normal movement.
#Checks the conditions and stores the result as True or False.
normal_move = (
    signal == "green" and
    pedestrian == "no"
)
```

```python
#Checks if an emergency vehicle can move when the signal is yellow.
#Checks the conditions and stores the result as True or False.
emergency_move = (
    signal == "yellow" and
    emergency == "yes"
)
```

```python
#Checks whether either normal or emergency movement is allowed and the emergency stop is not active.
#Checks the conditions and stores the result as True or False.
can_move = (
    (normal_move or emergency_move) and
    emergency_stop == "no"
)
```

```python
#Displays whether the vehicle can move (True or False).
print("Vehicle Can Move:", can_move)
```
