Task:

*   Accept a person's height in metres and weight in kilograms.
*   Calculate BMI and determine how much weight the person would need to gain or lose to reach a BMI of 22.

```python
#Asks the user to enter their height & weight, and converts it into a decimal number (float).
height = float(input("Enter height in metres: "))
weight = float(input("Enter weight in kg: "))

#Calculates BMI using the formula: Weight ÷ Height².
bmi = weight / (height ** 2)
```

```python
#Calculates the ideal weight using 22 × Height².
ideal_weight = 22 * (height ** 2)

#Finds the difference between the actual weight and ideal weight.
difference = weight - ideal_weight
```

```python
#displays the calculated BMI and ideal weight, rounded to 2 decimal places.
print("BMI:", round(bmi, 2))
print("Ideal Weight:", round(ideal_weight, 2))
```

```python
#Checks if the actual weight is more than the ideal weight and displays how much weight needs to be lost.
if difference > 0:
    print("Weight to lose:", round(difference, 2), "kg")

#Checks if the actual weight is less than the ideal weight and displays how much weight needs to be gained.
#abs() converts the difference into a positive value.
elif difference < 0:
    print("Weight to gain:", round(abs(difference), 2), "kg")

#Runs when the actual weight is exactly equal to the ideal weight and displays a message saying the person is at their ideal weight.
else:
    print("You are at the ideal weight.")
```
