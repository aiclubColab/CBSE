**Task:**

To write a Python program that:

*   Accepts the student's full name.
*   Accepts the student's year of birth.
*   Extracts the first three characters of the first name.
*   Extracts the last two digits of the birth year.
*   Combines them to create a username.

```python
#Takes the student's full name as input.
name = input("Enter your full name: ")

#Takes the student's birth year as input.
year = input("Enter your birth year: ")
```

```python
#Splits the name into words and selects the first name.
first_name = name.split()[0]

#Creates the username using the first three characters of the first name in lowercase and the last two digits of the birth year.
username = first_name[:3].lower() + year[-2:]
```

```python
#Displays the generated username.
print("Generated Username:", username)
```
