```python
#Write a Python program to create a digital wallet.
#For each transaction, ask the user whether the transaction is a credit or debit and update the wallet balance accordingly.

wallet = {
    "name": input("Enter your name: "),
    "balance": float(input("Enter initial balance: ")),
    "transactions": []
}

n = int(input("Enter number of transactions: "))

for i in range(n):
    transaction_type = input(
        "Enter transaction type (credit/debit): "
    ).lower()

    amount = float(input("Enter transaction amount: "))

    if transaction_type == "credit":
        wallet["balance"] += amount
        wallet["transactions"].append(
            ("Credit", amount)
        )

    elif transaction_type == "debit":
        wallet["balance"] -= amount
        wallet["transactions"].append(
            ("Debit", amount)
        )

    else:
        print("Invalid transaction type.")

print("\nWallet Details")
print("----------------")
print("Name:", wallet["name"])
print("Final Balance:", wallet["balance"])

print("Transactions:")
for transaction in wallet["transactions"]:
    print(transaction[0], "₹", transaction[1])
```
