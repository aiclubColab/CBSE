**Task:**

To create a simple digital wallet program that:
*   Accepts the user's name and initial balance.
*   Processes multiple credit and debit transactions.
*   Updates the wallet balance.
*   Stores and displays the transaction history.

```python
#Creates a dictionary to store the user's name, wallet balance, and transaction history.
wallet = {

    #Takes the user's name as input.
    "name": input("Enter your name: "),

    #Takes the initial wallet balance as a decimal number.
    "balance": float(input("Enter initial balance: ")),

    #Creates an empty list to store transactions.
    "transactions": []
}
```

```python
#Takes the number of transactions to be entered.
n = int(input("Enter number of transactions: "))
```

```python
#Repeats the loop n times for each transaction.
for i in range(n):

    #Takes the transaction type and converts it to lowercase.
    transaction_type = input(
        "Enter transaction type (credit/debit): "
    ).lower()

    #Takes the transaction amount as a decimal number.
    amount = float(input("Enter transaction amount: "))

    #Checks whether the transaction is a credit
    if transaction_type == "credit":

        #Adds the amount to the wallet balance.
        wallet["balance"] += amount

        #Records the credit transaction and amount in the transaction list.
        wallet["transactions"].append(
            ("Credit", amount)
        )

    #Checks whether the transaction is a debit.
    elif transaction_type == "debit":

        #Subtracts the amount from the wallet balance.
        wallet["balance"] -= amount

        #Records the debit transaction and amount in the transaction list.
        wallet["transactions"].append(
            ("Debit", amount)
        )

    #Runs when an invalid transaction type is entered.
    else:

        #Displays a message for an invalid transaction.
        print("Invalid transaction type.")
```

```python
#Displays the Wallet Details heading.
print("\nWallet Details")

#Displays a separator line.
print("----------------")

#Displays the wallet owner's name.
print("Name:", wallet["name"])

#Displays the final wallet balance.
print("Final Balance:", wallet["balance"])
```

```python
#Displays the Transactions heading.
print("Transactions:")

#Loops through all the recorded transactions.
for transaction in wallet["transactions"]:

    #Displays the transaction type and amount.
    print(transaction[0], "₹", transaction[1])
```
