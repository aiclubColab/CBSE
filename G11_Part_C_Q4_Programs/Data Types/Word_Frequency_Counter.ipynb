**Task:**

To write a Python program that:
*   Accepts a paragraph or sentence from the user.
*   Converts the text into lowercase.
*   Splits the paragraph into individual words.
*   Finds the frequency of each unique word.
*   Stores the word frequencies in a dictionary.
*   Displays each word and its frequency.

```python
#Takes a sentence as input and converts it to lowercase.
text = input("Enter a sentence: ").lower()
```

```python
#Splits the sentence into a list of individual words.
words = text.split()

#Creates a set containing only unique words.
unique_words = set(words)
```

```python
#Creates an empty dictionary to store each word and its frequency.
frequency = {}
```

```python
#Loops through each unique word.
for word in unique_words:

    #Counts how many times each word appears and stores the frequency in the dictionary.
    frequency[word] = words.count(word)
```

```python
#Displays the Word Frequency heading.
print("\nWord Frequency")
```

```python
#Loops through each word and its frequency in the dictionary.
for word, count in frequency.items():

    #Displays each word and the number of times it appears.
    print(word, ":", count)
```
