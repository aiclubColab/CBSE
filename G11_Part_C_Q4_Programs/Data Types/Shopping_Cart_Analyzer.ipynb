```python
'''
Write a Python program that allows the user to enter the names and prices of products in a shopping cart. The program should display:

All products and their prices
Total number of products
Total cost
Most expensive product and its price
'''

products = []
prices = []

n = int(input("Enter the number of products: "))

for i in range(n):
    product = input("Enter product name: ")
    price = float(input("Enter product price: "))

    products.append(product)
    prices.append(price)

print("\nShopping Cart")
print("----------------")

for i in range(len(products)):
    print(products[i], "₹", prices[i])

total = sum(prices)
highest_price = max(prices)
highest_index = prices.index(highest_price)

print("\nTotal number of products:", len(products))
print("Total cost: ₹", total)
print("Most expensive product:", products[highest_index])
print("Highest price: ₹", highest_price)
```
