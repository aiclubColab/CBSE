**Task:**

To write a Python program that:
*   Accepts product names and prices.
*   Stores the products and their prices.
*   Displays all products with their prices.
*   Finds the total number of products.
*   Calculates the total cost.
*   Finds the most expensive product and its price.

```python
#Creates an empty list to store product names.
products = []

#Creates an empty list to store product prices.
prices = []
```

```python
#Takes the number of products from the user.
n = int(input("Enter the number of products: "))
```

```python
#Repeats the loop n times to enter each product.
for i in range(n):

    #Takes the product name as input.
    product = input("Enter product name: ")

    #Takes the product price as a decimal number.
    price = float(input("Enter product price: "))

    #Adds the product name to the products list.
    products.append(product)

    #Adds the product price to the prices list.
    prices.append(price)
```

```python
#Displays the Shopping Cart heading.
print("\nShopping Cart")

#Displays a separator line.
print("----------------")
```

```python
#Loops through all the products using their index positions.
for i in range(len(products)):

    #Displays each product and its price.
    print(products[i], "₹", prices[i])
```

```python
#Calculates the total cost of all products.
total = sum(prices)

#Finds the highest price in the prices list.
highest_price = max(prices)

#Finds the index of the most expensive product.
highest_index = prices.index(highest_price)
```

```python
#Displays the total number of products.
print("\nTotal number of products:", len(products))

#Displays the total cost of the shopping cart.
print("Total cost: ₹", total)

#Displays the name of the most expensive product.
print("Most expensive product:", products[highest_index])

#Displays the highest product price.
print("Highest price: ₹", highest_price)
```
