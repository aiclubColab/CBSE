**Task:**

To write a Python program that:
*   Accepts the food items ordered by a group of students.
*   Stores all the food items in a list.
*   Searches for a particular food item.
*   Finds how many times that food item was ordered.
*   Displays the complete list of orders.
*   Displays the total number of orders.

```python
#Creates an empty list to store the food orders.
orders = []

#Takes the number of orders from the user as an integer.
n = int(input("Enter number of orders: "))
```

```python
#Repeats the loop n times to accept each order.
for i in range(n):

    #Takes a food item and converts it to lowercase.
    item = input("Enter food item: ").lower()

    #Adds the food item to the orders list.
    orders.append(item)
```

```python
#Takes the food item to search for and converts it to lowercase.
search = input("Enter item to search: ").lower()
```

```python
#Displays the complete list of orders.
print("All Orders:", orders)

#Counts and displays how many times the searched food item was ordered.
print(search, "was ordered", orders.count(search), "time(s)")

#Displays the total number of orders in the list.
print("Total Orders:", len(orders))
```
