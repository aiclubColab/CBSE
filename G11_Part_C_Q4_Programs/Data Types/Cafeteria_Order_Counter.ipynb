```python
#Accept food items ordered by students. Find how many times a particular food item was ordered.

orders = []

n = int(input("Enter number of orders: "))

for i in range(n):
    item = input("Enter food item: ").lower()
    orders.append(item)

search = input("Enter item to search: ").lower()

print("All Orders:", orders)
print(search, "was ordered", orders.count(search), "time(s)")
print("Total Orders:", len(orders))
```
