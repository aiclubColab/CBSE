**Task:**

To write a Python program that:

*   Accepts a number from the user.
*   Accepts a limit for the multiplication table.
*   Generates the multiplication table using a while loop.

```python
#Takes the number for which the multiplication table is required.
number = int(input("Enter number: "))

#Takes the limit up to which the table should be generated.
limit = int(input("Enter limit: "))
```

```python
#Initializes the counter i to 1.
i = 1
```

```python
#Repeats the loop while i is less than or equal to the limit.
while i <= limit:

    #Displays the multiplication result for the current value of i.
    print(number, "x", i, "=", number * i)

    #Increases i by 1 for the next multiplication.
    i += 1
```
