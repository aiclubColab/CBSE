**Task:**

To write a Python program that:
*   Accepts the type of vehicle (bike or car).
*   Accepts the number of hours parked.
*   Calculates the parking fee based on the vehicle type and parking duration.
*   Handles invalid vehicle types.

```python
#Takes the vehicle type and converts it to lowercase.
vehicle = input(
    "Enter vehicle type (bike/car): "
).lower()

#Takes the number of hours parked as an integer.
hours = int(input("Enter parking hours: "))
```

```python
#Checks if the vehicle is a bike.
if vehicle == "bike":

    #Checks if the bike is parked for 2 hours or less.
    if hours <= 2:

        #Sets the parking fee to ₹20.
        fee = 20

    #Checks if the bike is parked for 3 to 5 hours.
    elif hours <= 5:

        #Sets the parking fee to ₹40.
        fee = 40

    #Runs when the bike is parked for more than 5 hours.
    else:

        #Sets the parking fee to ₹60.
        fee = 60

#Checks if the vehicle is a car.
elif vehicle == "car":

    #Checks if the car is parked for 2 hours or less.
    if hours <= 2:

        #Sets the parking fee to ₹40.
        fee = 40

    #Checks if the car is parked for 3 to 5 hours.
    elif hours <= 5:

        #Sets the parking fee to ₹80.
        fee = 80

    #Runs when the car is parked for more than 5 hours.
    else:

        #Sets the parking fee to ₹120.
        fee = 120

#Runs when an invalid vehicle type is entered.
else:

    #Sets the parking fee to ₹0 for an invalid vehicle type.
    fee = 0

    #Displays a message indicating an invalid vehicle type.
    print("Invalid vehicle type.")
```

```python
#Displays the calculated parking fee.
print("Parking Fee: ₹", fee)
```
