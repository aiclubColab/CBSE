```python
#Calculate parking fees based on vehicle type and number of hours parked.

vehicle = input(
    "Enter vehicle type (bike/car): "
).lower()

hours = int(input("Enter parking hours: "))

if vehicle == "bike":

    if hours <= 2:
        fee = 20
    elif hours <= 5:
        fee = 40
    else:
        fee = 60

elif vehicle == "car":

    if hours <= 2:
        fee = 40
    elif hours <= 5:
        fee = 80
    else:
        fee = 120

else:
    fee = 0
    print("Invalid vehicle type.")

print("Parking Fee: ₹", fee)
```
