**Task:**

To write a Python program that:
*   Accepts the attendance status of 10 students.
*   Identifies students as Present (P) or Absent (A).
*   Counts the total number of present students.
*   Counts the total number of absent students.
*   Handles invalid entries.

```python
#Initializes the present student count to 0.
present = 0

#Initializes the absent student count to 0.
absent = 0
```

```python
A#Repeats the loop for 10 students, from 1 to 10.
for i in range(1, 11):

    #Takes the attendance status and converts it to uppercase.
    status = input(
        "Enter attendance for student " + str(i) + " (P/A): "
    ).upper()

    #Checks if the student is present and increases the present count by 1 if present.
    if status == "P":
        present += 1

    #Checks if the student is absent and increases the absent count by 1 if absent.
    elif status == "A":
        absent += 1

    #Runs when an invalid attendance entry is given.
    else:

        #Displays a message for an invalid entry.
        print("Invalid entry.")
```

```python
#Displays the total number of present students.
print("\nPresent:", present)

#Displays the total number of absent students.
print("Absent:", absent)
```
