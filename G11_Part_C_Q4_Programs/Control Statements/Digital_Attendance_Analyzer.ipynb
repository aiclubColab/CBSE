```python
#Accept attendance status for 10 students and count the number of students who are present and absent.

present = 0
absent = 0

for i in range(1, 11):
    status = input(
        "Enter attendance for student " + str(i) + " (P/A): "
    ).upper()

    if status == "P":
        present += 1

    elif status == "A":
        absent += 1

    else:
        print("Invalid entry.")

print("\nPresent:", present)
print("Absent:", absent)
```
