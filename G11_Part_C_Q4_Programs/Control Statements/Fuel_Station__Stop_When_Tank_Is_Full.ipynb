```python
#Add fuel to a tank. Invalid quantities should be skipped. Stop when the tank reaches 100 litres.

fuel = 0

while fuel < 100:

    amount = float(input("Enter fuel to add: "))

    if amount <= 0:
        print("Invalid quantity.")
        continue

    if fuel + amount > 100:
        print("Amount exceeds tank capacity.")
        continue

    fuel += amount

    print("Fuel level:", fuel, "litres")

    if fuel == 100:
        print("Tank is full.")
        break
```
