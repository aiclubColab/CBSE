**Task:**

To write a Python program that:
*   Accepts the amount of fuel to be added to a tank.
*   Skips invalid quantities such as 0 or negative values.
*   Prevents the fuel level from exceeding 100 litres.
*   Continues accepting fuel until the tank is full.
*   Uses continue to skip invalid entries and break to stop when the tank reaches 100 litres.

```python
#Initializes the fuel level to 0 litres.
fuel = 0
```

```python
#Repeats the loop until the tank reaches 100 litres.
while fuel < 100:

    #Takes the amount of fuel to be added as a decimal number.
    amount = float(input("Enter fuel to add: "))

    #Checks if the entered fuel quantity is zero or negative.
    if amount <= 0:

        #Displays a message for an invalid quantity.
        print("Invalid quantity.")

        #Skips the remaining code and starts the next loop iteration.
        continue

    #Checks if adding the fuel would exceed the tank capacity.
    if fuel + amount > 100:

        #Displays a message when the amount exceeds 100 litres.
        print("Amount exceeds tank capacity.")

        #Skips the current iteration and asks for another fuel amount.
        continue

    #Adds the valid fuel amount to the tank.
    fuel += amount

    #Displays the current fuel level.
    print("Fuel level:", fuel, "litres")

    #Checks if the tank has reached its full capacity.
    if fuel == 100:

        #Displays a message that the tank is full.
        print("Tank is full.")

        #Stops the loop when the tank reaches 100 litres.
        break
```
