**Task:**

To write a Python program that:
*   Records rainfall for 7 days.
*   Calculates the total rainfall.
*   Calculates the average rainfall.
*   Counts the number of heavy-rain days.
*   Consider rainfall above 50 mm as heavy rain.

```python
#Initializes the total rainfall to 0.
total = 0

#Initializes the count of heavy-rain days to 0.
heavy_days = 0
```

```python
#Repeats the loop for 7 days, from Day 1 to Day 7.
for day in range(1, 8):

    #Takes the rainfall amount for each day as a decimal number.
    rainfall = float(input(
        "Enter rainfall for day " + str(day) + " in mm: "
    ))

    #Adds the day's rainfall to the total rainfall.
    total += rainfall

    #Checks whether the rainfall is more than 50 mm.
    if rainfall > 50:

        #Increases the heavy-rain day count by 1.
        heavy_days += 1
```

```python
#Displays the total rainfall for the 7 days.
print("Total Rainfall:", total, "mm")

#Calculates and displays the average rainfall.
print("Average Rainfall:", total / 7, "mm")

#Displays the number of heavy-rain days.
print("Heavy Rain Days:", heavy_days)
```
