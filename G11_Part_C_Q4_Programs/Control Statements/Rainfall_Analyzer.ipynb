```python
#Record rainfall for 7 days and determine the total rainfall and number of heavy-rain days.

total = 0
heavy_days = 0

for day in range(1, 8):

    rainfall = float(input(
        "Enter rainfall for day " + str(day) + " in mm: "
    ))

    total += rainfall

    if rainfall > 50:
        heavy_days += 1

print("Total Rainfall:", total, "mm")
print("Average Rainfall:", total / 7, "mm")
print("Heavy Rain Days:", heavy_days)
```
