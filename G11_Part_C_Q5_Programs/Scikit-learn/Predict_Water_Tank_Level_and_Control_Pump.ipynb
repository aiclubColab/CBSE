**Task:**

To write a Python program that:
*   Uses Scikit-learn to create a machine learning model.
*   Uses the current water level and daily water usage as input features.
*   Predicts whether the tank has a Low or Normal water level.
*   Uses the K-Nearest Neighbors (KNN) classification algorithm.
*   Decides whether the water pump should be switched ON or OFF based on the prediction.
*   Displays the predicted tank status and the required action.

```python
#Imports the K-Nearest Neighbors (KNN) classification algorithm.
from sklearn.neighbors import KNeighborsClassifier
```

```python
#Stores the training data containing tank level and daily water usage.
#Each pair represents the tank level percentage and daily water usage.
X = [
    [20, 50],
    [25, 45],
    [30, 40],
    [35, 35],
    [60, 30],
    [70, 25],
    [80, 20],
    [90, 15]
]
```

```python
#Stores the corresponding tank status for each training example.
#Represents whether the tank has a Low or Normal water level.
y = [
    "Low", "Low", "Low", "Low",
    "Normal", "Normal", "Normal", "Normal"
]
```

```python
#Creates a KNN model that uses the 3 nearest examples to make a prediction.
model = KNeighborsClassifier(n_neighbors=3)

#Trains the model using the tank level, water usage, and status data.
model.fit(X, y)

#Takes the current tank level percentage from the user.
level = int(input("Enter tank level percentage: "))

#Takes the daily water usage from the user.
usage = int(input("Enter daily water usage: "))

#Uses the trained model to predict the tank status.
prediction = model.predict([[level, usage]])

#Displays the predicted Low or Normal tank status.
print("\nPredicted Tank Status:", prediction[0])
```

```python
#Checks if the predicted tank status is Low and displays the instruction to switch ON the pump when the tank level is Low.
if prediction[0] == "Low":
    print("Action: Switch ON the water pump.")

#Runs when the predicted tank status is Normal and displays the instruction to keep the pump OFF when the tank level is Normal.
else:
    print("Action: Keep the pump OFF.")
```
