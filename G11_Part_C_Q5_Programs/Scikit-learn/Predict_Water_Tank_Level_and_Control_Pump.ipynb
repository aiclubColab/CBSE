```python
'''
Predict whether a water tank has "Low" or "Normal" water level based on current water level and daily usage.
Then decide whether the pump should be switched on.
'''

from sklearn.neighbors import KNeighborsClassifier

X = [
    [20, 50],
    [25, 45],
    [30, 40],
    [35, 35],
    [60, 30],
    [70, 25],
    [80, 20],
    [90, 15]
]

y = [
    "Low", "Low", "Low", "Low",
    "Normal", "Normal", "Normal", "Normal"
]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

level = int(input("Enter tank level percentage: "))
usage = int(input("Enter daily water usage: "))

prediction = model.predict([[level, usage]])

print("\nPredicted Tank Status:", prediction[0])

if prediction[0] == "Low":
    print("Action: Switch ON the water pump.")
else:
    print("Action: Keep the pump OFF.")
```
