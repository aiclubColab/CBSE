**Task:**

To write a Python program that:

*   Uses Scikit-learn to create a machine learning model.
*   Uses the area of a house in square feet as the input.
*   Uses house price in lakhs as the target value.
*   Applies Linear Regression to learn the relationship between house area and price.
*   Predicts the approximate price of a new house based on its area.

```python
#Imports the Linear Regression machine learning algorithm.
from sklearn.linear_model import LinearRegression
```

```python
#Stores the house areas in square feet as training data.
area = [[500], [700], [900], [1100], [1300], [1500]]

#Stores the corresponding house prices in lakhs as training data.
price = [20, 28, 35, 42, 50, 58]
```

```python
#Creates a Linear Regression model.
model = LinearRegression()

#Trains the model using house area and price data.
model.fit(area, price)

#Takes the area of a new house from the user.
house_area = float(input("Enter house area in sq.ft: "))

#Uses the trained model to predict the approximate house price.
prediction = model.predict([[house_area]])
```

```python
#Displays the predicted house price in lakhs.
print("Predicted House Price:", prediction[0], "lakhs")
```
