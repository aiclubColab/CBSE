```python
'''
Predict whether a package should be delivered by Bike or Van based on distance and weight.
Also display the number of Bike and Van deliveries in the dataset.
'''

from sklearn.neighbors import KNeighborsClassifier

X = [
    [2, 1],
    [3, 2],
    [4, 2],
    [5, 3],
    [8, 6],
    [10, 8],
    [12, 10],
    [15, 12]
]

y = [
    "Bike",
    "Bike",
    "Bike",
    "Bike",
    "Van",
    "Van",
    "Van",
    "Van"
]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

distance = float(input("Enter delivery distance in km: "))
weight = float(input("Enter package weight in kg: "))

prediction = model.predict([[distance, weight]])

print("\nSuggested Vehicle:", prediction[0])

print("Bike Deliveries:", y.count("Bike"))
print("Van Deliveries:", y.count("Van"))
```
