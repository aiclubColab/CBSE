```python
'''
Create a simple machine learning model to predict whether a person belongs to a Low Risk or High Risk category based on age, BMI, and daily exercise.
Also display the predicted category.
'''

from sklearn.neighbors import KNeighborsClassifier

# Data: [Age, BMI, Exercise Hours]
X = [
    [20, 21, 2],
    [25, 22, 3],
    [30, 23, 2],
    [35, 24, 3],
    [40, 29, 1],
    [45, 31, 1],
    [50, 32, 0],
    [55, 34, 1]
]

y = [
    "Low Risk",
    "Low Risk",
    "Low Risk",
    "Low Risk",
    "High Risk",
    "High Risk",
    "High Risk",
    "High Risk"
]

model = KNeighborsClassifier(n_neighbors=3)

model.fit(X, y)

age = int(input("Enter age: "))
bmi = float(input("Enter BMI: "))
exercise = float(input("Enter daily exercise hours: "))

prediction = model.predict([[age, bmi, exercise]])

print("\nPredicted Category:", prediction[0])

print("Number of Low Risk Records:", y.count("Low Risk"))
print("Number of High Risk Records:", y.count("High Risk"))
```
