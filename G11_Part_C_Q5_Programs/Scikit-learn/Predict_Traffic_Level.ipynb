```python
#Predict whether traffic will be "Low", "Medium", or "High" based on the number of vehicles and time of day.

from sklearn.neighbors import KNeighborsClassifier

X = [
    [20, 8],
    [30, 9],
    [40, 10],
    [60, 8],
    [70, 9],
    [80, 10],
    [100, 8],
    [120, 9],
    [140, 10]
]

y = [
    "Low",
    "Low",
    "Medium",
    "Medium",
    "Medium",
    "Medium",
    "High",
    "High",
    "High"
]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

vehicles = int(input("Enter number of vehicles: "))
hour = int(input("Enter hour of the day: "))

prediction = model.predict([[vehicles, hour]])

print("Predicted Traffic:", prediction[0])
```
