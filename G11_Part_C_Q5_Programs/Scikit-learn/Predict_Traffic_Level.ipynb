**Task:**

To write a Python program that:
*   Uses Scikit-learn to create a machine learning model.
*   Uses the number of vehicles and time of day as input features.
*   Predicts whether traffic will be Low, Medium, or High.
*   Uses the K-Nearest Neighbors (KNN) classification algorithm.
*   Accepts new traffic information from the user.
*   Displays the predicted traffic level.

```python
#Imports the K-Nearest Neighbors (KNN) classification algorithm.
from sklearn.neighbors import KNeighborsClassifier
```

```python
#Stores the training data containing the number of vehicles and time of day.
#Each pair represents the number of vehicles and hour of the day.
X = [
    [20, 8],
    [30, 9],
    [40, 10],
    [60, 8],
    [70, 9],
    [80, 10],
    [100, 8],
    [120, 9],
    [140, 10]
]
```

```python
#Stores the corresponding traffic category for each training example.
#Represents the traffic level for each training example.
y = [
    "Low",
    "Low",
    "Medium",
    "Medium",
    "Medium",
    "Medium",
    "High",
    "High",
    "High"
]
```

```python
#Creates a KNN model that uses the 3 nearest examples to make a prediction.
model = KNeighborsClassifier(n_neighbors=3)

#Trains the model using the vehicle, time, and traffic data.
model.fit(X, y)

#Takes the number of vehicles from the user.
vehicles = int(input("Enter number of vehicles: "))

#Takes the hour of the day from the user.
hour = int(input("Enter hour of the day: "))

#Uses the trained model to predict the traffic level.
prediction = model.predict([[vehicles, hour]])

#Displays the predicted traffic category.
print("Predicted Traffic:", prediction[0])
```
