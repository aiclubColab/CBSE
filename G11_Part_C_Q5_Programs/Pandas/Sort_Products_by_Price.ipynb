```python
#Enter product names and prices. Display the products from lowest to highest price.

import pandas as pd

products = []
prices = []

for i in range(6):
    products.append(input("Enter product name: "))
    prices.append(float(input("Enter price: ")))

df = pd.DataFrame({
    "Product": products,
    "Price": prices
})

sorted_df = df.sort_values("Price")

print("\nProducts sorted by price:")
print(sorted_df)
```
