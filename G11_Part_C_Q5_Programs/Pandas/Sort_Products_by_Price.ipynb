**Task:**

To write a Python program that:
*   Accepts the names and prices of 6 products.
*   Stores the product details in a Pandas DataFrame.
*   Sorts the products from lowest to highest price.
*   Displays the sorted product list.

```python
#Imports the Pandas library and gives it the shorter name pd.
import pandas as pd

#Creates an empty list to store product names.
products = []

#Creates an empty list to store product prices.
prices = []
```

```python
#Repeats the loop to enter details for 6 products.
for i in range(6):

    #Takes the product name and adds it to the products list.
    products.append(input("Enter product name: "))

    #Takes the product price and adds it to the prices list.
    prices.append(float(input("Enter price: ")))
```

```python
#Creates a DataFrame using the product names and prices.
df = pd.DataFrame({

    #Creates the Product column using the product names.
    "Product": products,

    #Creates the Price column using the product prices.
    "Price": prices
})

#Sorts the DataFrame by price from lowest to highest.
sorted_df = df.sort_values("Price")
```

```python
#Displays the heading for the sorted products.
print("\nProducts sorted by price:")

#Displays the products in ascending order of price.
print(sorted_df)
```
