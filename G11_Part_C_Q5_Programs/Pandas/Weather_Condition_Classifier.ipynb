**Task:**

To write a Python program that:
*   Accepts rainfall measurements for 5 days.
*   Stores the day and rainfall data in a Pandas DataFrame.
*   Classifies each day based on the amount of rainfall as:

            Dry – rainfall is 0 mm.
            Moderate – rainfall is greater than 0 mm and up to 20 mm.
            Heavy – rainfall is more than 20 mm.
*   Adds the weather condition to the DataFrame.
*   Displays the complete weather report.

```python
#Imports the Pandas library and gives it the shorter name pd.
import pandas as pd

#Creates an empty list to store the days.
days = []

#Creates an empty list to store rainfall measurements.
rainfall = []
```

```python
#Repeats the loop to enter rainfall details for 5 days.
for i in range(5):

    #Takes the day name as input.
    day = input("Enter day: ")

    #Takes the rainfall measurement in millimetres.
    rain = float(input("Enter rainfall in mm: "))

    #Adds the day to the days list.
    days.append(day)

    #Adds the rainfall value to the rainfall list.
    rainfall.append(rain)
```

```python
#Creates a DataFrame using the days and rainfall values.
df = pd.DataFrame({

    #Creates the Day column using the day names.
    "Day": days,

    #Creates the Rainfall column using the rainfall measurements.
    "Rainfall": rainfall
})
```

```python
#Defines a function to classify rainfall based on its amount.
def classify(rain):

    #Checks if the rainfall is 0 mm and returns "Dry" when there is no rainfall.
    if rain == 0:
        return "Dry"

    #Checks if the rainfall is 20 mm or less and returns "Moderate" for rainfall up to 20 mm.
    elif rain <= 20:
        return "Moderate"

    #Handles rainfall above 20 mm and returns "Heavy" for rainfall above 20 mm.
    else:
        return "Heavy"
```

```python
#Applies the classify function to rainfall values and creates a Condition column.
df["Condition"] = df["Rainfall"].apply(classify)

#Displays the heading for the weather report.
print("\nWeather Report:")

#Displays the days, rainfall measurements, and classifications.
print(df)
```
