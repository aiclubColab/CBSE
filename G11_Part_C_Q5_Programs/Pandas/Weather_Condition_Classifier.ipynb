```python
#Store rainfall measurements and classify each day as "Dry", "Moderate", or "Heavy".

import pandas as pd

days = []
rainfall = []

for i in range(5):
    day = input("Enter day: ")
    rain = float(input("Enter rainfall in mm: "))

    days.append(day)
    rainfall.append(rain)

df = pd.DataFrame({
    "Day": days,
    "Rainfall": rainfall
})

def classify(rain):
    if rain == 0:
        return "Dry"
    elif rain <= 20:
        return "Moderate"
    else:
        return "Heavy"

df["Condition"] = df["Rainfall"].apply(classify)

print("\nWeather Report:")
print(df)
```
