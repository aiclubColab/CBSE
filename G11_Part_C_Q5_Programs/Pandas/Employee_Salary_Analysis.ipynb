**Task:**

To write a Python program that:
*   Accepts details of 5 employees.
*   Records each employee's name, department, and salary.
*   Creates a Pandas DataFrame using the collected information.
*   Displays all employee details.
*   Displays employees earning more than ₹50,000.

```python
#Imports the Pandas library and gives it the shorter name pd.
import pandas as pd

#Creates an empty list to store employee names.
names = []

#Creates an empty list to store employee departments.
departments = []

#Creates an empty list to store employee salaries.
salaries = []
```

```python
#Repeats the loop to enter details for 5 employees.
for i in range(5):

    #Takes the employee's name as input.
    name = input("Enter employee name: ")

    #Takes the employee's department as input.
    department = input("Enter department: ")

    #Takes the employee's salary as input.
    salary = float(input("Enter salary: "))

    #Adds the employee's name to the names list.
    names.append(name)

    #Adds the department to the departments list.
    departments.append(department)

    #Adds the salary to the salaries list.
    salaries.append(salary)
```

```python
#Creates a DataFrame using the names, departments, and salaries.
df = pd.DataFrame({

    #Creates the Name column in the DataFrame.
    "Name": names,

    #Creates the Department column in the DataFrame.
    "Department": departments,

    #Creates the Salary column in the DataFrame.
    "Salary": salaries
})
```

```python
#Displays the heading for employee data.
print("\nEmployee Data:")

#Displays the complete employee DataFrame.
print(df)

#Displays the heading for employees earning more than ₹50,000.
print("\nEmployees earning above ₹50,000:")

#Filters and displays employees whose salary is greater than ₹50,000.
print(df[df["Salary"] > 50000])
```
