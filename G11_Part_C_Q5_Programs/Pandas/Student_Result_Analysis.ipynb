**Task:**

To write a Python program that:
*   Accepts the names and marks of 6 students.
*   Creates a Pandas DataFrame containing student details.
*   Determines whether each student has Passed or Failed.
*   Adds a new Result column to the DataFrame.
*   Displays the student names, marks, and results.

```python
#Imports the Pandas library and gives it the shorter name pd.
import pandas as pd

#Creates an empty list to store student names.
names = []

#Creates an empty list to store student marks.
marks = []
```

```python
#Repeats the loop to enter details for 6 students.
for i in range(6):

    #Takes the student's name as input.
    name = input("Enter student name: ")

    #Takes the student's marks as input.
    mark = float(input("Enter marks: "))

    #Adds the student's name to the names list.
    names.append(name)

    #Adds the student's marks to the marks list.
    marks.append(mark)
```

```python
#Creates a DataFrame using the names and marks.
df = pd.DataFrame({

    #Creates the Name column using the student names.
    "Name": names,

    #Creates the Marks column using the student marks.
    "Marks": marks
})

#Creates a new Result column based on the marks.
df["Result"] = df["Marks"].apply(

    #Marks the student "Pass" if marks are 40 or more; otherwise, marks "Fail".
    lambda x: "Pass" if x >= 40 else "Fail"
)
```

```python
#Displays the student names, marks, and results.
print(df)
```
