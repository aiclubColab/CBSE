```python
'''
Enter the names and marks of students.
Add a new column called Result.
Students scoring 40 or more should be marked "Pass".
'''

import pandas as pd

names = []
marks = []

for i in range(6):
    name = input("Enter student name: ")
    mark = float(input("Enter marks: "))

    names.append(name)
    marks.append(mark)

df = pd.DataFrame({
    "Name": names,
    "Marks": marks
})

df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print(df)
```
