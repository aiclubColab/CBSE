```python
'''
Create a DataFrame containing book details.
Allow the user to add a new book, remove an existing book, or modify the price of a book.
'''

import pandas as pd

df = pd.DataFrame({
    "Book ID": [101, 102, 103, 104],
    "Book": ["Python Basics", "Data Science", "AI Fundamentals", "Web Design"],
    "Price": [350, 500, 450, 400]
})

print("Library Book Data:")
print(df)

print("\n1. Add a Book")
print("2. Remove a Book")
print("3. Modify Book Price")

choice = int(input("Enter your choice: "))

if choice == 1:
    book_id = int(input("Enter Book ID: "))
    book = input("Enter Book Name: ")
    price = float(input("Enter Book Price: "))

    new_book = pd.DataFrame({
        "Book ID": [book_id],
        "Book": [book],
        "Price": [price]
    })

    df = pd.concat([df, new_book], ignore_index=True)

elif choice == 2:
    book_id = int(input("Enter Book ID to remove: "))

    df = df[df["Book ID"] != book_id]
    df = df.reset_index(drop=True)

elif choice == 3:
    book_id = int(input("Enter Book ID to modify: "))
    new_price = float(input("Enter new price: "))

    df.loc[df["Book ID"] == book_id, "Price"] = new_price

else:
    print("Invalid choice!")

print("\nUpdated Library Data:")
print(df)
```
