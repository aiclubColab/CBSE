**Task:**

To write a Python program that:
*   Creates a DataFrame containing book details.
*   Displays the available books in a library.
*   Allows the user to add a new book.
*   Allows the user to remove an existing book.
*   Allows the user to modify the price of a book.
*   Displays the updated library data.

```python
#Imports the Pandas library and gives it the shorter name pd.
import pandas as pd
```

```python
#Creates a DataFrame containing the book details.
df = pd.DataFrame({

    #Creates the Book ID column with the book identification numbers.
    "Book ID": [101, 102, 103, 104],

    #Creates the Book column with book names.
    "Book": ["Python Basics", "Data Science", "AI Fundamentals", "Web Design"],

    #Creates the Price column with book prices.
    "Price": [350, 500, 450, 400]
})
```

```python
#Displays the heading for the library data.
print("Library Book Data:")

#Displays the book details in the DataFrame.
print(df)

#Displays the option to add a book.
print("\n1. Add a Book")

#Displays the option to remove a book.
print("2. Remove a Book")

#Displays the option to modify a book's price.
print("3. Modify Book Price")
```

```python
#Takes the user's menu choice as an integer.
choice = int(input("Enter your choice: "))
```

```python
#Checks if the user wants to add a new book.
if choice == 1:

    #Takes the Book ID of the new book.
    book_id = int(input("Enter Book ID: "))

    #Takes the name of the new book.
    book = input("Enter Book Name: ")

    #Takes the price of the new book.
    price = float(input("Enter Book Price: "))

    #Creates a new DataFrame containing the new book's details.
    new_book = pd.DataFrame({
        "Book ID": [book_id],
        "Book": [book],
        "Price": [price]
    })

    #Adds the new book to the existing DataFrame and resets the index.
    df = pd.concat([df, new_book], ignore_index=True)

#Checks if the user wants to remove a book.
elif choice == 2:

    #Takes the Book ID to be removed.
    book_id = int(input("Enter Book ID to remove: "))

    #Removes the row with the specified Book ID.
    df = df[df["Book ID"] != book_id]

    #Resets the DataFrame index after removing the book.
    df = df.reset_index(drop=True)

#Checks if the user wants to modify a book's price.
elif choice == 3:

    #Takes the Book ID whose price needs to be changed.
    book_id = int(input("Enter Book ID to modify: "))

    #Takes the new price for the book.
    new_price = float(input("Enter new price: "))

    #Finds the specified book and updates its price.
    df.loc[df["Book ID"] == book_id, "Price"] = new_price

#Runs when the user enters an invalid choice.
else:

    #Displays an invalid choice message.
    print("Invalid choice!")
```

```python
#Displays the heading for the updated data.
print("\nUpdated Library Data:")

#Displays the updated library DataFrame.
print(df)
```
