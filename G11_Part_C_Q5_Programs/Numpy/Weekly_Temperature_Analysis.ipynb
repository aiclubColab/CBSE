```python
#Record the temperature for 7 days from the user and find the average temperature, highest temperature, lowest temperature, and number of days above 30°C.

import numpy as np

temperature = []

for day in range(1, 8):
    temp = float(input("Enter temperature for day " + str(day) + ": "))
    temperature.append(temp)

temperature = np.array(temperature)

average = np.mean(temperature)
highest = np.max(temperature)
lowest = np.min(temperature)
above_30 = np.sum(temperature > 30)

print("Temperatures:", temperature)
print("Average Temperature:", average)
print("Highest Temperature:", highest)
print("Lowest Temperature:", lowest)
print("Days above 30°C:", above_30)
```
