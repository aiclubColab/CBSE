**Task:**

To write a Python program that:
*   Record the temperature for 7 days from the user.
*   Stores the temperatures in a NumPy array.
*   Find the average temperature.
*   Find the highest temperature.
*   Find the lowest temperature.
*   Counts the number of days with temperature above 30°C.

```python
#Imports the NumPy library and gives it the shorter name np.
import numpy as np

#Creates an empty list to store the temperatures.
temperature = []
```

```python
#Repeats the loop for 7 days.
for day in range(1, 8):

    #Takes the temperature for each day from the user.
    temp = float(input("Enter temperature for day " + str(day) + ": "))

    #Adds the temperature to the temperature list.
    temperature.append(temp)
```

```python
#Converts the list into a NumPy array.
temperature = np.array(temperature)
```

```python
#Calculates the average temperature.
average = np.mean(temperature)

#Finds the highest temperature.
highest = np.max(temperature)

#Finds the lowest temperature.
lowest = np.min(temperature)

#Counts the number of days with temperature above 30°C.
above_30 = np.sum(temperature > 30)
```

```python
#Displays the temperatures for all 7 days.
print("Temperatures:", temperature)

#Displays the average temperature.
print("Average Temperature:", average)

#Displays the highest temperature.
print("Highest Temperature:", highest)

#Displays the lowest temperature.
print("Lowest Temperature:", lowest)

#Displays the number of days above 30°C.
print("Days above 30°C:", above_30)
```
