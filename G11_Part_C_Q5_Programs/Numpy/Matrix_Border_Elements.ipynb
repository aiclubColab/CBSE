**Task**

To write a Python program that:
*   Accepts values for a 4 × 4 matrix.
*   Stores the values in a NumPy array.
*   Displays the complete matrix.
*   Extracts and displays only the border elements of the matrix.
*   Displays the top row, bottom row, left column, and right column separately.

```python
#Imports the NumPy library and gives it the shorter name np.
import numpy as np

#Creates an empty list to store the matrix.
matrix = []
```

```python
#Repeats the loop for 4 rows.
for i in range(4):

    #Creates an empty list to store values in the current row.
    row = []

    #Repeats the loop for 4 columns.
    for j in range(4):

        #Takes an integer value from the user.
        value = int(input("Enter value: "))

        #Adds the value to the current row.
        row.append(value)

    #Adds the completed row to the matrix.
    matrix.append(row)
```

```python
#Converts the list into a 4 × 4 NumPy array.
matrix = np.array(matrix)
```

```python
#Displays the heading for the matrix.
print("Matrix:")

#Displays the 4 × 4 matrix.
print(matrix)

#Displays the top row of the matrix.
print("Top Row:", matrix[0])

#Displays the bottom row of the matrix.
print("Bottom Row:", matrix[-1])

#Displays the left column of the matrix.
print("Left Column:", matrix[:, 0])

#Displays the right column of the matrix.
print("Right Column:", matrix[:, -1])
```
