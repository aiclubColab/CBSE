```python
#Enter a 4 × 4 matrix and display only its border elements.

import numpy as np

matrix = []

for i in range(4):
    row = []
    for j in range(4):
        value = int(input("Enter value: "))
        row.append(value)
    matrix.append(row)

matrix = np.array(matrix)

print("Matrix:")
print(matrix)

print("Top Row:", matrix[0])
print("Bottom Row:", matrix[-1])
print("Left Column:", matrix[:, 0])
print("Right Column:", matrix[:, -1])
```
