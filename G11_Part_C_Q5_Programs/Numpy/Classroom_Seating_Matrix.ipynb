```python
'''
Create a 4 × 5 matrix representing a classroom.
Enter 1 if a seat is occupied and 0 if it is empty.
Find the total number of occupied and empty seats.
'''

import numpy as np

seats = []
00

for i in range(4):
    row = []

    for j in range(5):
        value = int(input(
            "Enter 1 for occupied, 0 for empty - Row " +
            str(i + 1) + ", Seat " + str(j + 1) + ": "
        ))
        row.append(value)

    seats.append(row)

seats = np.array(seats)

print("Seating Arrangement:")
print(seats)

print("Occupied Seats:", np.sum(seats))
print("Empty Seats:", np.sum(seats == 0))
```
