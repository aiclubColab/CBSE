**Task:**

To write a Python program that:
*   Creates a 4 × 5 matrix representing a classroom seating arrangement.
*   Accepts 1 if a seat is occupied and 0 if a seat is empty.
*   Displays the seating arrangement.
*   Finds the total number of occupied seats.
*   Finds the total number of empty seats.

```python
#Imports the NumPy library and gives it the shorter name np.
import numpy as np

#Creates an empty list to store the seating arrangement.
seats = []
```

```python
#Repeats the loop for 4 rows.
for i in range(4):

    #Creates an empty list to store seats in the current row.
    row = []

    #Repeats the loop for 5 seats in each row.
    for j in range(5):

        #Takes 1 for an occupied seat or 0 for an empty seat.
        value = int(input(
            "Enter 1 for occupied, 0 for empty - Row " +
            str(i + 1) + ", Seat " + str(j + 1) + ": "
        ))

        #Adds the seat value to the current row.
        row.append(value)

    #Adds the completed row to the seating arrangement.
    seats.append(row)
```

```python
#Converts the list into a 4 × 5 NumPy array.
seats = np.array(seats)
```

```python
#Displays the heading for the seating arrangement.
print("Seating Arrangement:")

#Displays the 4 × 5 seating matrix.
print(seats)

#Calculates and displays the total number of occupied seats.
print("Occupied Seats:", np.sum(seats))

#Counts and displays the total number of empty seats.
print("Empty Seats:", np.sum(seats == 0))
```
