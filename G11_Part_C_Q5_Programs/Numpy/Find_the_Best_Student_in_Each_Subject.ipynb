```python
#Enter marks of 5 students in 3 subjects and find the student who scored the highest in each subject.

import numpy as np

marks = []

for i in range(5):
    row = []

    for j in range(3):
        mark = int(input(
            "Enter marks of student " + str(i + 1) +
            " in subject " + str(j + 1) + ": "
        ))
        row.append(mark)

    marks.append(row)

marks = np.array(marks)

best_students = np.argmax(marks, axis=0)

print("Marks:")
print(marks)

print("Best student in Subject 1:", best_students[0] + 1)
print("Best student in Subject 2:", best_students[1] + 1)
print("Best student in Subject 3:", best_students[2] + 1)
```
