**Task:**

To write a Python program that:
*   Accepts marks of 5 students in 3 subjects.
*   Stores the marks in a NumPy array.
*   Finds the student who scored the highest marks in each subject.
*   Displays the marks and the best student for each subject.

```python
#Imports the NumPy library and gives it the shorter name np.
import numpy as np

#Creates an empty list to store the marks.
marks = []
```

```python
#Repeats the loop for 5 students.
for i in range(5):

    #Creates an empty list to store marks for the current student.
    row = []

    #Repeats the loop for 3 subjects.
    for j in range(3):

        #Takes the marks of a student in a subject.
        mark = int(input(
            "Enter marks of student " + str(i + 1) +
            " in subject " + str(j + 1) + ": "
        ))

        #Adds the mark to the current student's row.
        row.append(mark)

    #Adds the completed row to the marks list.
    marks.append(row)
```

```python
#Converts the list into a 5 × 3 NumPy array.
marks = np.array(marks)

#Finds the student with the highest marks in each subject.
best_students = np.argmax(marks, axis=0)
```

```python
#Displays the heading for the marks.
print("Marks:")

#Displays the marks of all 5 students.
print(marks)

#Displays the student number with the highest marks in Subject 1.
print("Best student in Subject 1:", best_students[0] + 1)

#Displays the student number with the highest marks in Subject 2.
print("Best student in Subject 2:", best_students[1] + 1)

#Displays the student number with the highest marks in Subject 3.
print("Best student in Subject 3:", best_students[2] + 1)
```
