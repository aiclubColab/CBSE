```python
'''
Enter the number of passengers boarding a bus at 8 stops.
Find the maximum number of passengers boarding at one stop and the stops where more than 20 passengers boarded.
'''

import numpy as np

passengers = []

for i in range(8):
    count = int(input("Enter passengers at stop " + str(i + 1) + ": "))
    passengers.append(count)

passengers = np.array(passengers)

busy_stops = passengers[passengers > 20]

print("Passengers:", passengers)
print("Maximum Passengers:", np.max(passengers))
print("Stops with more than 20 passengers:", busy_stops)
```
