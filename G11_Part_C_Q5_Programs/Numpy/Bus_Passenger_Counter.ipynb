**Task:**

To write a Python program that:
*   Accepts the number of passengers boarding a bus at 8 stops.
*   Stores the passenger counts in a NumPy array.
*   Finds the maximum number of passengers boarding at one stop.
*   Identifies the passenger counts from stops where more than 20 passengers boarded.

```python
#Imports the NumPy library and gives it the shorter name np.
import numpy as np

#Creates an empty list to store passenger counts.
passengers = []
```

```python
#Repeats the loop for 8 bus stops.
for i in range(8):

    #Takes the number of passengers boarding at each stop.
    count = int(input("Enter passengers at stop " + str(i + 1) + ": "))

    #Adds the passenger count to the passengers list.
    passengers.append(count)
```

```python
#Converts the list into a NumPy array.
passengers = np.array(passengers)

#Selects the passenger counts from stops where more than 20 passengers boarded.
busy_stops = passengers[passengers > 20]
```

```python
#Displays the passenger counts for all 8 stops.
print("Passengers:", passengers)

#Finds and displays the maximum number of passengers at any stop.
print("Maximum Passengers:", np.max(passengers))

#Displays the passenger counts for stops with more than 20 passengers.
print("Stops with more than 20 passengers:", busy_stops)
```
