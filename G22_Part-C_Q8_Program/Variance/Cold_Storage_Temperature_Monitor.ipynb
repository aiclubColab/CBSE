**Task**

A cold storage facility stores temperature-sensitive products. The temperature is recorded 10 times during the day.

Write a Python program to:

* Accept 10 temperature readings from the user.
* Calculate the mean temperature.
* Calculate the variance of the temperature readings.
* Use the variance to determine the stability of the storage temperature:

      Variance ≤ 1 → Stable
      Variance > 1 and ≤ 4 → Moderately Unstable
      Variance > 4 → Highly Unstable
*If the temperature is highly unstable, identify the readings that are farthest from the mean.
*Display a warning if the temperature condition is highly unstable.

```python
# Import the math module to use mathematical functions
import math

# Create an empty list to store temperature readings
temperatures = []
```

```python
# Display a message asking the user to enter temperature readings
print("Enter temperature readings for 10 time periods:")

# Repeat the loop for 10 time periods
for i in range(10):

    # Ask the user to enter the temperature for the current time period in °C
    temp = float(input("Time period " + str(i + 1) + ": "))

    # Add the entered temperature to the list
    temperatures.append(temp)
```

```python
# Calculate the mean (average) temperature
mean = sum(temperatures) / len(temperatures)

# Calculate the variance using the squared differences from the mean
variance = sum((x - mean) ** 2 for x in temperatures) / len(temperatures)
```

```python
# Display a heading for the temperature analysis
print("\n--- Temperature Analysis ---")

# Display the mean temperature, rounded to 2 decimal places
print("Mean Temperature:", round(mean, 2), "°C")

# Display the variance, rounded to 2 decimal places
print("Variance:", round(variance, 2))
```

```python
# Check if the variance is 1 or less
if variance <= 1:

    # Display that the temperature is stable
    print("Temperature Condition: Stable")

# Check if the variance is between 1 and 4
elif variance <= 4:

    # Display that the temperature is moderately unstable
    print("Temperature Condition: Moderately Unstable")

# If the variance is greater than 4
else:

    # Display that the temperature is highly unstable
    print("Temperature Condition: Highly Unstable")

    # Create an empty list to store the deviations from the mean
    deviations = []

    # Go through each temperature reading
    for i in range(len(temperatures)):

        # Calculate the absolute difference between the temperature and the mean
        deviation = abs(temperatures[i] - mean)

        # Add the deviation to the list
        deviations.append(deviation)

    # Find the largest deviation from the mean
    largest_deviation = max(deviations)

    # Display a warning about high temperature variation
    print("\nWARNING: Temperature variation is high!")

    # Display a heading for the readings farthest from the mean
    print("Readings farthest from the mean:")

    # Go through each temperature reading again
    for i in range(len(temperatures)):

        # Check if the current deviation is equal to the largest deviation
        if deviations[i] == largest_deviation:

            # Display the time period and temperature reading
            print("Time period", i + 1,
                  ":", temperatures[i], "°C")
```
