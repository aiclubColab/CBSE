**Task**

A restaurant records the number of customers arriving during 10 different one-hour periods.

Write a Python program to analyze the customer data and help the restaurant manager decide whether additional staff may be required.

The program should:

* Accept the number of customers for 10 one-hour periods.
* Calculate the mean number of customers.
* Calculate the variance of customer arrivals.
* Classify the customer flow based on variance:

      Variance ≤ 25 → Stable
      Variance > 25 and ≤ 100 → Moderately Variable
      Variance > 100 → Highly Variable
* If the customer flow is highly variable, identify the busiest time periods.
* Assume that one staff member can handle 20 customers per hour.
* Calculate the number of staff required during the busiest period.
* Compare it with the restaurant's current number of staff and determine whether additional staff are required.

```python
# Create an empty list to store the number of customers for each one-hour period
customers = []
```

```python
# Display a message asking the user to enter customer counts
print("Enter the number of customers for 10 one-hour periods:")

# Repeat the loop for 10 one-hour periods
for i in range(10):

    # Ask the user to enter the number of customers for the current hour
    count = int(input("Hour " + str(i + 1) + ": "))

    # Add the customer count to the list
    customers.append(count)
```

```python
# Calculate the mean (average) number of customers per hour
mean = sum(customers) / len(customers)

# Calculate the variance using the squared differences from the mean
variance = sum((x - mean) ** 2 for x in customers) / len(customers)

# Find the highest number of customers in any one-hour period
busiest_customers = max(customers)
```

```python
# Display a heading for the restaurant analysis
print("\n--- Restaurant Analysis ---")

# Display the average number of customers per hour
print("Average customers per hour:", round(mean, 2))

# Display the variance of the customer counts
print("Variance:", round(variance, 2))
```

```python
# Check if the variance is 25 or less
if variance <= 25:

    # Display that customer flow is stable
    print("Customer Flow: Stable")

# Check if the variance is between 25 and 100
elif variance <= 100:

    # Display that customer flow is moderately variable
    print("Customer Flow: Moderately Variable")

# If the variance is greater than 100
else:

    # Display that customer flow is highly variable
    print("Customer Flow: Highly Variable")
```

```python
# Display a heading for the busiest period
print("\nBusiest Period:")

# Go through each hour to find the busiest period
for i in range(len(customers)):

    # Check if the current hour has the highest number of customers
    if customers[i] == busiest_customers:

        # Display the hour and number of customers
        print("Hour", i + 1, ":", customers[i], "customers")
```

```python
# Ask the user to enter the current number of staff members
current_staff = int(input("\nEnter the current number of staff members: "))

# One staff member can handle 20 customers per hour
customers_per_staff = 20

# Calculate the number of staff required during the busiest period
# The calculation rounds up to ensure enough staff are available
required_staff = (busiest_customers + customers_per_staff - 1) // customers_per_staff
```

```python
# Display a heading for staff requirements
print("\n--- Staff Requirement ---")

# Display the current number of staff members
print("Current staff:", current_staff)

# Display the number of staff required during the busiest period
print("Staff required during busiest period:", required_staff)
```

```python
# Check if more staff members are required
if required_staff > current_staff:

    # Calculate the number of additional staff members needed
    additional_staff = required_staff - current_staff

    # Display the number of additional staff required
    print("Additional staff required:", additional_staff)

# If the current staff is enough
else:

    # Display that no additional staff members are required
    print("No additional staff are required.")
```
