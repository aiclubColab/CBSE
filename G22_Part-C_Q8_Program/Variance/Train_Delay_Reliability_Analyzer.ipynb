**Task**

A railway passenger wants to choose the more reliable train between Train A and Train B. The arrival delay of both trains is recorded for 10 different days.

Write a Python program to compare the two trains using variance.

The program should:

* Accept the arrival delay, in minutes, for 10 days for Train A.
* Accept the arrival delay for 10 days for Train B.
* Calculate the mean delay for both trains.
* Calculate the variance of the delays for both trains.
* Compare the variances.
* Recommend the train with the lower variance as the more reliable train.
* Display the difference between the two variances.
* If both trains have the same variance, report that they have equal reliability.

```python
# Create an empty list to store delay times for Train A
train_a = []

# Create an empty list to store delay times for Train B
train_b = []
```

```python
# Ask the user to enter arrival delays for Train A
print("Enter arrival delays for Train A (in minutes):")

# Repeat the input process 10 times for 10 days
for i in range(10):

    # Get the delay time for the current day
    delay = float(input("Day " + str(i + 1) + ": "))

    # Add the delay time to the Train A list
    train_a.append(delay)
```

```python
# Ask the user to enter arrival delays for Train B
print("\nEnter arrival delays for Train B (in minutes):")

# Repeat the input process 10 times for 10 days
for i in range(10):

    # Get the delay time for the current day
    delay = float(input("Day " + str(i + 1) + ": "))

    # Add the delay time to the Train B list
    train_b.append(delay)
```

```python
# Calculate the mean (average) delay for Train A
mean_a = sum(train_a) / len(train_a)

# Calculate the mean (average) delay for Train B
mean_b = sum(train_b) / len(train_b)


# Calculate variance for Train A
# (x - mean_a) finds how far each delay is from the average
# ** 2 squares the difference
# sum() adds all the squared differences
# / len(train_a) gives the average of these squared differences
variance_a = sum((x - mean_a) ** 2 for x in train_a) / len(train_a)


# Calculate variance for Train B
variance_b = sum((x - mean_b) ** 2 for x in train_b) / len(train_b)
```

```python
# Display the train reliability analysis
print("\n--- Train Reliability Analysis ---")

# Display results for Train A
print("\nTrain A")

# Display the mean delay rounded to 2 decimal places
print("Mean delay:", round(mean_a, 2), "minutes")

# Display the variance rounded to 2 decimal places
print("Variance:", round(variance_a, 2))

# Display results for Train B
print("\nTrain B")

# Display the mean delay rounded to 2 decimal places
print("Mean delay:", round(mean_b, 2), "minutes")

# Display the variance rounded to 2 decimal places
print("Variance:", round(variance_b, 2))
```

```python
# Calculate the absolute difference between the two variances
# abs() makes sure the difference is always positive
difference = abs(variance_a - variance_b)

# Display the difference in variance
print("\nDifference in variance:", round(difference, 2))
```

```python
# Compare the variances to decide which train is more reliable
# Lower variance means the delays are more consistent
if variance_a < variance_b:

    # Train A has lower variation in delays
    print("Recommendation: Train A is more reliable.")

    # Display the reason for the recommendation
    print("Reason: Train A has lower variation in arrival delays.")


# Check if Train B has lower variance
elif variance_b < variance_a:

    # Train B has lower variation in delays
    print("Recommendation: Train B is more reliable.")

    # Display the reason for the recommendation
    print("Reason: Train B has lower variation in arrival delays.")


# If both variances are equal
else:

    # Both trains have the same variation in delays
    print("Recommendation: Both trains have equal reliability.")
```
