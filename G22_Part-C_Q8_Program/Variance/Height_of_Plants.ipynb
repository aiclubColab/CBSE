**Task**

A botanist wants to determine which of two fertilizers, Fertilizer P and Fertilizer Q, produces more consistent plant growth. The height (in cm) of plants treated with each fertilizer is recorded after 10 weeks (one plant measured per week, or 10 different plants — same batch, same conditions).

Write a Python program to compare the two fertilizers using variance.

The program should:

* Accept the plant height (in cm) for 10 plants/weeks grown using Fertilizer P.
* Accept the plant height (in cm) for 10 plants/weeks grown using Fertilizer Q.
* Calculate the mean height for both fertilizers.
* Calculate the variance of the heights for both fertilizers.
* Compare the variances.
* Recommend the fertilizer with the lower variance as producing more consistent growth.
* Display the difference between the two variances.
* If both fertilizers have the same variance, report that they produce equally consistent growth.

```python
# Create an empty list to store plant heights for Fertilizer P
fertilizer_p = []

# Create an empty list to store plant heights for Fertilizer Q
fertilizer_q = []
```

```python
# Input plant heights for Fertilizer P
print("Enter plant heights for Fertilizer P:")

# Repeat the loop 10 times to enter heights of 10 plants
for i in range(10):

    # Ask the user to enter the height of the plant
    # str(i + 1) displays the plant number starting from 1
    # float() converts the entered value into a decimal number
    height = float(input("Plant " + str(i + 1) + ": "))

    # Add the entered height to the Fertilizer P list
    fertilizer_p.append(height)
```

```python
# Input plant heights for Fertilizer Q
print("\nEnter plant heights for Fertilizer Q:")

# Repeat the loop 10 times to enter heights of 10 plants
for i in range(10):

    # Ask the user to enter the height of the plant
    height = float(input("Plant " + str(i + 1) + ": "))

    # Add the entered height to the Fertilizer Q list
    fertilizer_q.append(height)
```

```python
# Calculate the mean height for Fertilizer P
# sum() adds all the heights and len() gives the number of plants
mean_p = sum(fertilizer_p) / len(fertilizer_p)

# Calculate the mean height for Fertilizer Q
mean_q = sum(fertilizer_q) / len(fertilizer_q)
```

```python
# Calculate the variance for Fertilizer P
# (x - mean_p) finds the difference between each height and the mean
# ** 2 squares the difference
# sum() adds all the squared differences
# Divide by the number of plants to get the variance
variance_p = sum((x - mean_p) ** 2 for x in fertilizer_p) / len(fertilizer_p)

# Calculate the variance for Fertilizer Q
variance_q = sum((x - mean_q) ** 2 for x in fertilizer_q) / len(fertilizer_q)

# Find the difference between the two variances
# abs() gives the positive value of the difference
difference = abs(variance_p - variance_q)
```

```python
# Display the results
print("\n--- Results ---")

# Display the mean height for Fertilizer P
# round() displays the value up to 2 decimal places
print("Mean height with Fertilizer P:", round(mean_p, 2), "cm")

# Display the mean height for Fertilizer Q
print("Mean height with Fertilizer Q:", round(mean_q, 2), "cm")


# Display the variance for Fertilizer P
print("Variance for Fertilizer P:", round(variance_p, 2))

# Display the variance for Fertilizer Q
print("Variance for Fertilizer Q:", round(variance_q, 2))

# Display the difference between the two variances
print("Difference between variances:", round(difference, 2))
```

```python
# Compare the two variances
if variance_p < variance_q:

    # If Fertilizer P has a lower variance,
    # its plant heights are more consistent
    print("Fertilizer P produces more consistent plant growth.")

elif variance_q < variance_p:

    # If Fertilizer Q has a lower variance,
    # its plant heights are more consistent
    print("Fertilizer Q produces more consistent plant growth.")

else:

    # If both variances are equal,
    # both fertilizers have the same consistency
    print("Both fertilizers produce equally consistent plant growth.")
```
