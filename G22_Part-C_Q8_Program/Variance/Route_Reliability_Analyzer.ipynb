**Task**

A person travels to work using two different routes: Route A and Route B. The travel time for each route is recorded over 10 different days.

Write a Python program to compare the reliability of the two routes.

The program should:

* Accept 10 travel-time readings for Route A.
* Accept 10 travel-time readings for Route B.
* Calculate the mean travel time for both routes.
* Calculate the variance of travel time for both routes.
* Compare the variances.
* Recommend the route with the lower variance as the more reliable route.
* Also display the difference between the two variances.
* If both routes have the same variance, report that they have equal reliability.

```python
# Create an empty list to store travel times for Route A
route_a = []

# Create an empty list to store travel times for Route B
route_b = []
```

```python
# Display a message asking the user to enter travel times for Route A
print("Enter travel times for Route A (in minutes):")

# Repeat the loop for 10 days
for i in range(10):

    # Ask the user to enter the travel time for Route A on the current day
    time = float(input("Day " + str(i + 1) + ": "))

    # Add the travel time to the Route A list
    route_a.append(time)
```

```python
# Display a message asking the user to enter travel times for Route B
print("\nEnter travel times for Route B (in minutes):")

# Repeat the loop for 10 days
for i in range(10):

    # Ask the user to enter the travel time for Route B on the current day
    time = float(input("Day " + str(i + 1) + ": "))

    # Add the travel time to the Route B list
    route_b.append(time)
```

```python
# Calculate the mean (average) travel time for Route A
mean_a = sum(route_a) / len(route_a)

# Calculate the mean (average) travel time for Route B
mean_b = sum(route_b) / len(route_b)

# Calculate the variance of travel times for Route A
variance_a = sum((x - mean_a) ** 2 for x in route_a) / len(route_a)

# Calculate the variance of travel times for Route B
variance_b = sum((x - mean_b) ** 2 for x in route_b) / len(route_b)
```

```python
# Display a heading for the route analysis
print("\n--- Route Analysis ---")

# Display the results for Route A
print("\nRoute A")

# Display the mean travel time for Route A
print("Mean travel time:", round(mean_a, 2), "minutes")

# Display the variance for Route A
print("Variance:", round(variance_a, 2))

# Display the results for Route B
print("\nRoute B")

# Display the mean travel time for Route B
print("Mean travel time:", round(mean_b, 2), "minutes")

# Display the variance for Route B
print("Variance:", round(variance_b, 2))
```

```python
# Calculate the absolute difference between the two variances
difference = abs(variance_a - variance_b)

# Display the difference in variance between the two routes
print("\nDifference in variance:", round(difference, 2))
```

```python
# Check if Route A has a lower variance than Route B
if variance_a < variance_b:

    # Recommend Route A because it has less variation in travel time
    print("Recommendation: Route A is more reliable.")

# Check if Route B has a lower variance than Route A
elif variance_b < variance_a:

    # Recommend Route B because it has less variation in travel time
    print("Recommendation: Route B is more reliable.")

# If both routes have the same variance
else:

    # State that both routes have equal reliability
    print("Recommendation: Both routes have equal reliability.")
```
