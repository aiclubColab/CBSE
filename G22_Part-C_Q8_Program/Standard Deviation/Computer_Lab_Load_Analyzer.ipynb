**Task:**

A school has a computer lab with a fixed number of computers. The lab administrator records the number of computers being used during 10 different time periods.

Write a Python program to:

* Store the number of computers in use during the 10 time periods.
* Calculate the average number of computers in use.
* Calculate the standard deviation.
* Calculate the expected peak usage using:

**Expected Peak Usage = Mean + Standard Deviation**

* Ask the user to enter the total number of computers available in the lab.
* Compare the expected peak usage with the available computers.
* If the expected peak usage is greater than the available computers, calculate and display the number of additional computers needed.
* Otherwise, display that the existing computers are sufficient.

```python
# Import the math module to use mathematical functions
import math

# Create an empty list to store the number of computers in use
usage = []
```

```python
# Display a message asking the user to enter computer usage
print("Enter the number of computers in use during 10 time periods:")

# Repeat the loop for 10 time periods
for i in range(10):

    # Ask the user to enter the number of computers being used
    computers = int(input("Time period " + str(i + 1) + ": "))

    # Add the number of computers in use to the list
    usage.append(computers)
```

```python
# Calculate the mean (average) number of computers in use
mean = sum(usage) / len(usage)

# Calculate the variance using the squared differences from the mean
variance = sum((x - mean) ** 2 for x in usage) / len(usage)

# Calculate the standard deviation by finding the square root of variance
sd = math.sqrt(variance)

# Estimate the expected peak usage by adding standard deviation to the mean
expected_peak = mean + sd
```

```python
# Ask the user to enter the total number of computers available
available = int(input("\nEnter the total number of computers available: "))
```

```python
# Display a heading for the computer lab analysis
print("\n--- Computer Lab Analysis ---")

# Display the computer usage during the 10 time periods
print("Usage:", usage)

# Display the average computer usage, rounded to 2 decimal places
print("Average Usage:", round(mean, 2))

# Display the standard deviation, rounded to 2 decimal places
print("Standard Deviation:", round(sd, 2))

# Display the expected peak computer usage
print("Expected Peak Usage:", round(expected_peak, 2))

# Display the total number of computers available
print("Available Computers:", available)
```

```python
# Check if the expected peak usage is greater than the available computers
if expected_peak > available:

    # Calculate the additional computers needed
    # math.ceil() rounds the value up to the next whole number
    additional = math.ceil(expected_peak - available)

    # Display the number of additional computers needed
    print("\nAdditional computers needed:", additional)

    # Display a warning about insufficient computers during peak usage
    print("The lab may not have enough computers during peak usage.")

# If the available computers are enough for the expected peak usage
else:

    # Calculate the number of computers available as a buffer
    extra = available - math.ceil(expected_peak)

    # Display a message that the existing computers are sufficient
    print("\nExisting computers are sufficient.")

    # Display the number of extra computers available as a buffer
    print("Computers available as buffer:", extra)
```
