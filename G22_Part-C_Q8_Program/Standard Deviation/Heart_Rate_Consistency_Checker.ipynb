**Task:**

A healthcare worker records a patient's heart rate at 10 different times during a monitoring period.

Write a Python program to:

* Store the 10 heart rate readings in a list.
* Calculate the mean heart rate.
* Calculate the standard deviation.
* Classify the variation in the readings:

      SD < 5 → Low variation
      SD from 5 to 10 → Moderate variation
      SD > 10 → High variation
* If the variation is high, display an alert message.

```python
# Import the math module to use mathematical functions
import math

# Create an empty list to store heart rate readings
heart_rate = []
```

```python
# Display a message asking the user to enter heart rate readings
print("Enter heart rate readings (in beats per minute):")

# Repeat the loop for 10 heart rate readings
for i in range(10):

    # Ask the user to enter the current heart rate in beats per minute
    reading = float(input("Reading " + str(i + 1) + ": "))

    # Add the entered heart rate reading to the list
    heart_rate.append(reading)
```

```python
# Calculate the mean (average) heart rate
mean = sum(heart_rate) / len(heart_rate)

# Calculate the variance using the squared differences from the mean
variance = sum((x - mean) ** 2 for x in heart_rate) / len(heart_rate)

# Calculate the standard deviation by finding the square root of variance
sd = math.sqrt(variance)
```

```python
# Display a heading for the heart rate report
print("\n--- Heart Rate Report ---")

# Display all the heart rate readings
print("Heart Rate Readings:", heart_rate)

# Display the mean heart rate, rounded to 2 decimal places
print("Mean Heart Rate:", round(mean, 2), "bpm")

# Display the standard deviation, rounded to 2 decimal places
print("Standard Deviation:", round(sd, 2))
```

```python
# Check if the standard deviation is less than 5
if sd < 5:

    # Display that the variation in readings is low
    print("Variation: Low")

# Check if the standard deviation is between 5 and 10
elif sd <= 10:

    # Display that the variation in readings is moderate
    print("Variation: Moderate")

# If the standard deviation is greater than 10
else:

    # Display that the variation in readings is high
    print("Variation: High")

    # Display an alert about the high variation
    print("\nAlert: Heart-rate readings show high variation.")

    # Suggest that the readings may need further monitoring
    print("Further monitoring may be required.")
```
