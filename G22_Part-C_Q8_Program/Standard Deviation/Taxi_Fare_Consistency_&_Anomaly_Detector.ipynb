**Task:**

A taxi company records the fares charged for 10 different trips. Write a Python program to analyze the fares and identify trips with unusually high or low fares.

The program should:

* Store the fares of 10 trips in a list.
* Calculate the average fare.
* Calculate the standard deviation.
* Calculate the:

        Lower Limit = Mean − Standard Deviation
        Upper Limit = Mean + Standard Deviation
* Classify each trip as:

        Normal → Fare is within the limits
        Unusual → Fare is below the lower limit or above the upper limit
* Display the number of normal and unusual trips.
* Display the trip numbers and fares that are identified as unusual.

```python
# Import the math module to use mathematical functions
import math

# Create an empty list to store the fares of 10 taxi trips
fares = []
```

```python
# Display a message asking the user to enter taxi fares
print("Enter the fares for 10 taxi trips:")

# Repeat the loop for 10 taxi trips
for i in range(10):

    # Ask the user to enter the fare for the current trip in rupees
    fare = float(input("Trip " + str(i + 1) + ": ₹"))

    # Add the entered fare to the list
    fares.append(fare)
```

```python
# Calculate the mean (average) taxi fare
mean = sum(fares) / len(fares)

# Calculate the variance using the squared differences from the mean
variance = sum((x - mean) ** 2 for x in fares) / len(fares)

# Calculate the standard deviation by finding the square root of variance
sd = math.sqrt(variance)
```

```python
# Calculate the lower limit using mean minus standard deviation
lower_limit = mean - sd

# Calculate the upper limit using mean plus standard deviation
upper_limit = mean + sd
```

```python
# Display a heading for the taxi fare analysis
print("\n--- Taxi Fare Analysis ---")

# Display the average taxi fare, rounded to 2 decimal places
print("Average Fare: ₹", round(mean, 2))

# Display the standard deviation, rounded to 2 decimal places
print("Standard Deviation: ₹", round(sd, 2))

# Display the lower limit, rounded to 2 decimal places
print("Lower Limit: ₹", round(lower_limit, 2))

# Display the upper limit, rounded to 2 decimal places
print("Upper Limit: ₹", round(upper_limit, 2))
```

```python
# Create a variable to count normal trips
normal_count = 0

# Create a variable to count unusual trips
unusual_count = 0
```

```python
# Display a heading for classifying the trips
print("\n--- Trip Classification ---")

# Go through each taxi fare in the list
for i in range(len(fares)):

    # Check if the fare is below the lower limit or above the upper limit
    if fares[i] < lower_limit or fares[i] > upper_limit:

        # Display the trip as unusual
        print("Trip", i + 1, "→ Unusual (₹", fares[i], ")")

        # Increase the unusual trip count by 1
        unusual_count += 1

    # If the fare is within the lower and upper limits
    else:

        # Display the trip as normal
        print("Trip", i + 1, "→ Normal (₹", fares[i], ")")

        # Increase the normal trip count by 1
        normal_count += 1
```

```python
# Display a heading for the final summary
print("\n--- Summary ---")

# Display the total number of normal trips
print("Normal Trips:", normal_count)

# Display the total number of unusual trips
print("Unusual Trips:", unusual_count)
```
