**Task:**

A student records their expenses for 10 days. Write a program to:

* Accept the expenses.
* Calculate the average daily expense.
* Calculate the standard deviation.
* Find the highest and lowest expenses.
* Determine whether the student's spending is consistent.

```python
# Import the math module to use mathematical functions
import math

# Create an empty list to store expenses for 10 days
expenses = []
```

```python
# Display a message asking the user to enter daily expenses
print("Enter expenses for 10 days:")

# Repeat the loop for 10 days
for i in range(10):

    # Ask the user to enter the expense for the current day in rupees
    amount = float(input("Day " + str(i + 1) + ": ₹"))

    # Add the entered expense to the list
    expenses.append(amount)
```

```python
# Calculate the mean (average) daily expense
mean = sum(expenses) / len(expenses)

# Calculate the variance using the squared differences from the mean
variance = sum((x - mean) ** 2 for x in expenses) / len(expenses)

# Calculate the standard deviation by finding the square root of variance
sd = math.sqrt(variance)
```

```python
# Display the average daily expense, rounded to 2 decimal places
print("\nAverage Daily Expense: ₹", round(mean, 2))

# Display the highest expense
print("Highest Expense: ₹", max(expenses))

# Display the lowest expense
print("Lowest Expense: ₹", min(expenses))

# Display the standard deviation, rounded to 2 decimal places
print("Standard Deviation: ₹", round(sd, 2))
```

```python
# Check if the standard deviation is less than ₹100
if sd < 100:

    # Display this message if spending is relatively consistent
    print("Spending is relatively consistent.")

# If the standard deviation is ₹100 or more
else:

    # Display this message if spending varies considerably
    print("Spending varies considerably.")
```
