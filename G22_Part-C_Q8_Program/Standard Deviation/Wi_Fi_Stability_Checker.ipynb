**Task:**

A student wants to check how stable their Wi-Fi connection is throughout the day. Write a Python program to record the internet speed at 10 different times during the day.

The program should:

* Store the 10 internet speed readings in a list.
* Calculate the mean internet speed.
* Calculate the standard deviation.
* Classify the Wi-Fi connection based on the standard deviation:

      SD < 5 → Stable
      SD from 5 to 10 → Moderately Stable
      SD > 10 → Unstable
* Display the mean speed, standard deviation, and stability classification.

```python
# Import the math module to use mathematical functions
import math

# Create an empty list to store internet speed readings
speeds = []
```

```python
# Display a message asking the user to enter internet speed readings
print("Enter internet speed readings in Mbps:")

# Repeat the loop for 10 speed readings
for i in range(10):

    # Ask the user to enter the current internet speed in Mbps
    speed = float(input("Reading " + str(i + 1) + ": "))

    # Add the entered speed to the list
    speeds.append(speed)
```

```python
# Calculate the mean (average) internet speed
mean = sum(speeds) / len(speeds)

# Calculate the variance using the squared differences from the mean
variance = sum((x - mean) ** 2 for x in speeds) / len(speeds)

# Calculate the standard deviation by finding the square root of variance
sd = math.sqrt(variance)
```

```python
# Display a heading for the Wi-Fi stability report
print("\n--- Wi-Fi Stability Report ---")

# Display all the internet speed readings
print("Internet Speeds:", speeds)

# Display the mean internet speed, rounded to 2 decimal places
print("Mean Speed:", round(mean, 2), "Mbps")

# Display the standard deviation, rounded to 2 decimal places
print("Standard Deviation:", round(sd, 2))
```

```python
# Check if the standard deviation is less than 5 Mbps
if sd < 5:

    # Display that the Wi-Fi connection is stable
    print("Wi-Fi Status: Stable")

# Check if the standard deviation is between 5 and 10 Mbps
elif sd <= 10:

    # Display that the Wi-Fi connection is moderately stable
    print("Wi-Fi Status: Moderately Stable")

# If the standard deviation is greater than 10 Mbps
else:

    # Display that the Wi-Fi connection is unstable
    print("Wi-Fi Status: Unstable")
```
