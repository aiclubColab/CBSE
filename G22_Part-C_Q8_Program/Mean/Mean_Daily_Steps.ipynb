**Task:**

Write a python program that

*   Accepts the number of steps walked for 7 days.
*   Calculates the mean number of steps.
*   Identifies whether the student achieved an average of at least 8,000 steps.

```python
#Create an empty list to store the number of steps for each day
steps = []
```

```python
# Repeat the loop for 7 days
for day in range(1, 8):

    # Ask the user to enter the number of steps for the current day
    value = int(input("Enter steps for day " + str(day) + ": "))

    # Add the entered steps to the list
    steps.append(value)
```

```python
# Calculate the mean (average) number of steps
mean = sum(steps) / len(steps)

# Display the calculated mean number of steps
print("Mean Steps:", mean)
```

```python
# Check whether the mean number of steps is at least 8,000
if mean >= 8000:

    # Display this message if the target is achieved
    print("Target achieved!")

# If the mean is less than 8,000
else:

    # Display this message if the target is not achieved
    print("Target not achieved.")
```
