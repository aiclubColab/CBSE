**Task:**

Write a Python program that

* Accepts the marks scored by two students in 5 subjects.
* Calculates the mean marks of each student.
* Compares their mean marks.
* Displays the name of the student with the higher mean.

```python
# Create an empty list to store marks of Student 1
student1 = []

# Create an empty list to store marks of Student 2
student2 = []
```

```python
# Display a message to enter marks for Student 1
print("Enter marks for Student 1:")

# Repeat the loop for 5 subjects
for i in range(1, 6):

    # Ask the user to enter marks for the current subject
    mark = float(input("Enter marks for subject " + str(i) + ": "))

    # Add the entered mark to Student 1's list
    student1.append(mark)
```

```python
# Display a message to enter marks for Student 2
print("\nEnter marks for Student 2:")

# Repeat the loop for 5 subjects
for i in range(1, 6):

    # Ask the user to enter marks for the current subject
    mark = float(input("Enter marks for subject " + str(i) + ": "))

    # Add the entered mark to Student 2's list
    student2.append(mark)
```

```python
# Calculate the mean marks of Student 1
mean1 = sum(student1) / len(student1)

# Calculate the mean marks of Student 2
mean2 = sum(student2) / len(student2)

# Display the mean marks of Student 1
print("\nMean of Student 1:", mean1)

# Display the mean marks of Student 2
print("Mean of Student 2:", mean2)
```

```python
# Check if Student 1 has a higher mean than Student 2
if mean1 > mean2:

    # Display a message if Student 1 has the higher mean
    print("Student 1 has the higher mean.")

# Check if Student 2 has a higher mean than Student 1
elif mean2 > mean1:

    # Display a message if Student 2 has the higher mean
    print("Student 2 has the higher mean.")

# If both means are equal
else:

    # Display a message that both students have the same mean
    print("Both students have the same mean.")
```
