**Task:**

Write a Python program that

* Accepts the number of points scored by a player in 8 matches.
* Determines the player's usual scoring performance.
* Identifies matches where the player performed at least 20% better than usual.
* Identifies matches where the player performed at least 20% below usual.
* Displays whether the player's performance was "Consistent" or "Inconsistent" based on the number of unusual matches.

```python
# Create an empty list to store the scores of 8 matches
scores = []
```

```python

# Display a message asking the user to enter points scored in each match
print("Enter points scored in 8 matches:")

# Repeat the loop for 8 matches
for i in range(1, 9):

    # Ask the user to enter the score for the current match
    score = float(input("Enter score for match " + str(i) + ": "))

    # Add the entered score to the list
    scores.append(score)
```

```python
# Calculate the usual scoring performance by finding the mean (average) score
usual_score = sum(scores) / len(scores)

# Display the usual scoring performance
print("\nUsual Scoring Performance:", usual_score)
```

```python
# Create a variable to count matches with significantly better performance
better_matches = 0

# Create a variable to count matches with significantly lower performance
lower_matches = 0
```

```python
# Display a heading for matches with significantly better performance
print("\nMatches with significantly better performance:")

# Repeat the loop for each of the 8 matches
for i in range(8):

    # Check if the score is at least 20% higher than the usual score
    if scores[i] >= usual_score * 1.20:

        # Display the match number and the score
        print("Match", i + 1, ":", scores[i])

        # Increase the count of better matches by 1
        better_matches += 1
```

```python
# Display a heading for matches with significantly lower performance
print("\nMatches with significantly lower performance:")

# Repeat the loop for each of the 8 matches
for i in range(8):

    # Check if the score is at least 20% lower than the usual score
    if scores[i] <= usual_score * 0.80:

        # Display the match number and the score
        print("Match", i + 1, ":", scores[i])

        # Increase the count of lower matches by 1
        lower_matches += 1
```

```python
# Add the number of better and lower matches to find total unusual matches
unusual_matches = better_matches + lower_matches
```

```python
# Display the number of matches with better performance
print("\nBetter Matches:", better_matches)

# Display the number of matches with lower performance
print("Lower Matches:", lower_matches)

# Display the total number of unusual matches
print("Total Unusual Matches:", unusual_matches)
```

```python
# Check if there are 2 or fewer unusual matches
if unusual_matches <= 2:

    # Display that the player's performance is consistent
    print("Performance: Consistent")

# If there are more than 2 unusual matches
else:

    # Display that the player's performance is inconsistent
    print("Performance: Inconsistent")
```
