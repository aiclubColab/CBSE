**Task:**

Write a Python program that

* Accepts the number of hours a student studied for 7 days.
* Calculates the mean study hours.
* Identifies the days on which the student studied more than the mean.
* Displays "Consistent" if the student studied above the mean on at least 4 days.

```python
# Create an empty list to store study hours for each day
hours = []
```

```python
# Repeat the loop for 7 days
for day in range(1, 8):

    # Ask the user to enter the study hours for the current day
    value = float(input("Enter study hours for day " + str(day) + ": "))

    # Add the entered study hours to the list
    hours.append(value)
```

```python
# Calculate the mean (average) study hours
mean = sum(hours) / len(hours)

# Create a variable to count the number of days above the mean
above_mean = 0
```

```python
# Go through each study hour value in the list
for value in hours:

    # Check whether the study hours are greater than the mean
    if value > mean:

        # Increase the count by 1
        above_mean += 1
```

```python
# Display the calculated mean study hours
print("Mean Study Hours:", mean)

# Display the number of days with study hours above the mean
print("Days above mean:", above_mean)
```

```python
# Check whether at least 4 days are above the mean
if above_mean >= 4:

    # Display this message if the student is consistent
    print("Consistent")

# If fewer than 4 days are above the mean
else:

    # Display this message if improvement is needed
    print("Needs improvement")
```
