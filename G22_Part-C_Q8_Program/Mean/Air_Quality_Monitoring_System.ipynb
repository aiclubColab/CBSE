**Task:**

Write a Python program that

* Accepts AQI values for 7 days.
* Calculates the mean AQI.
* Counts the days above and below the mean.
* Finds the highest AQI value.
* Classifies the week's air quality based on the mean AQI.

**AQI** stands for Air Quality Index. It is a numerical value used to describe how clean or polluted the air is and how the air quality may affect people's health.

📊 **AQI Range**

AQI Value	Air Quality

* 0–50	🟢 Good – Air quality is satisfactory

* 51–100	🟡 Moderate – Acceptable, but some people may be sensitive

* 101–200	🟠 Poor – May affect people's health

* 201–300	🔴 Very Poor – Health effects are more likely

* 301–400	🟣 Severe – Serious health effects possible

* 401–500	⚫ Hazardous – Health risk for everyone

In simple terms:

👉 Lower AQI = Cleaner air

👉 Higher AQI = More polluted air and greater health risk

```python
# Create an empty list to store AQI values for each day
aqi = []
```

```python
# Repeat the loop for 7 days
for day in range(1, 8):

    # Ask the user to enter the Air Quality Index (AQI) for the current day
    value = int(input("Enter AQI for day " + str(day) + ": "))

    # Add the entered AQI value to the list
    aqi.append(value)
```

```python
# Calculate the mean (average) AQI
mean = sum(aqi) / len(aqi)
```

```python
# Create a variable to count the number of days above the mean
above = 0

# Create a variable to count the number of days below the mean
below = 0
```

```python
# Go through each AQI value in the list
for value in aqi:

    # Check whether the AQI is greater than the mean
    if value > mean:

        # Increase the count of days above the mean by 1
        above += 1

    # Check whether the AQI is less than the mean
    elif value < mean:

        # Increase the count of days below the mean by 1
        below += 1
```

```python
# Display the calculated mean AQI
print("Mean AQI:", mean)

# Display the highest AQI value
print("Highest AQI:", max(aqi))

# Display the number of days with AQI above the mean
print("Days Above Mean:", above)

# Display the number of days with AQI below the mean
print("Days Below Mean:", below)
```

```python
# Check if the mean AQI is 50 or less
if mean <= 50:

    # Display that the air quality is good
    print("Air Quality: Good")

# Check if the mean AQI is 100 or less
elif mean <= 100:

    # Display that the air quality is moderate
    print("Air Quality: Moderate")

# If the mean AQI is greater than 100
else:

    # Display that the air quality is poor
    print("Air Quality: Poor")
```
