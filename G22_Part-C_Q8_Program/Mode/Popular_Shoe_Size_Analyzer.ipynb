**Task:**

Write a Python program that

* Accepts the shoe sizes purchased by 11 customers.
* Determines the most commonly purchased shoe size.
* Displays the shoe size that should be stocked in greater quantity.
* Counts how many customers purchased this size.

```python
# Create an empty list to store the shoe sizes purchased by 11 customers
shoe_sizes = []
```

```python
# Display a message asking the user to enter shoe sizes
print("Enter the shoe sizes purchased by 11 customers:")

# Repeat the loop for 11 customers
for i in range(1, 12):

    # Ask the user to enter the shoe size of the current customer
    size = int(input("Enter shoe size of customer " + str(i) + ": "))

    # Add the entered shoe size to the list
    shoe_sizes.append(size)
```

```python
# Assume the first shoe size is the most commonly purchased size
popular_size = shoe_sizes[0]

# Count how many times the first shoe size appears in the list
maximum_count = shoe_sizes.count(popular_size)
```

```python
# Go through each shoe size in the list
for size in shoe_sizes:

    # Count how many times the current shoe size appears in the list
    count = shoe_sizes.count(size)

    # Check if the current shoe size has a higher count
    # than the maximum count found so far
    if count > maximum_count:

        # Update the maximum count
        maximum_count = count

        # Update the most commonly purchased shoe size
        popular_size = size
```

```python
# Display all the shoe sizes entered by the customers
print("\nShoe Sizes Purchased:", shoe_sizes)

# Display the most commonly purchased shoe size
print("Most Commonly Purchased Shoe Size:", popular_size)

# Display the number of customers who purchased this shoe size
print("Number of Customers:", maximum_count)

# Display a suggestion to stock more of the popular shoe size
print("Stock this shoe size in greater quantity.")
```
