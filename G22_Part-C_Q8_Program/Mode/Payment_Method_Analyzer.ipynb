**Task:**

Write a Python program that

* Accepts the payment method used for 13 transactions (UPI, Card, Cash, or Net Banking).
* Determines the most frequently used payment method.
* Displays the preferred payment method.
* Calculates how many transactions used that method.

```python
# Create an empty list to store the payment methods used in 13 transactions
payment_methods = []
```

```python
# Display a message asking the user to enter payment methods
print("Enter the payment method for 13 transactions:")

# Display the available payment method options
print("Options: UPI, Card, Cash, Net Banking")

# Repeat the loop for 13 transactions
for i in range(1, 14):

    # Ask the user to enter the payment method for the current transaction
    method = input("Enter payment method for transaction " + str(i) + ": ")

    # Add the entered payment method to the list
    payment_methods.append(method)
```

```python
# Assume the first payment method is the most frequently used method
preferred_method = payment_methods[0]

# Count how many times the first payment method appears in the list
maximum_count = payment_methods.count(preferred_method)
```

```python
# Go through each payment method in the list
for method in payment_methods:

    # Count how many times the current payment method appears in the list
    count = payment_methods.count(method)

    # Check if the current method has a higher count
    # than the maximum count found so far
    if count > maximum_count:

        # Update the maximum count
        maximum_count = count

        # Update the preferred payment method
        preferred_method = method
```

```python
# Display all the payment methods entered
print("\nPayment Methods:", payment_methods)

# Display the most frequently used payment method
print("Preferred Payment Method:", preferred_method)

# Display the number of transactions using the preferred method
print("Number of Transactions:", maximum_count)
```
