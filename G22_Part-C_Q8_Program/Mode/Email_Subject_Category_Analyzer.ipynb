**Task:**

Write a Python program that:

* Accepts the category of 15 incoming emails from the following categories:

           Work
           Shopping
           Social
           Education
           Other
* Determines the most frequently occurring email category.
* Displays the dominant email category.
* Calculates the percentage of emails belonging to the dominant category.
* Estimates how many emails of this category are expected in the next 100 emails, based on the observed pattern.

```python
# Create an empty list to store the categories of 15 incoming emails
emails = []
```

```python
# Display a message asking the user to enter email categories
print("Enter the category of 15 incoming emails:")

# Display the available email category options
print("Categories: Work, Shopping, Social, Education, Other")

# Repeat the loop for 15 emails
for i in range(1, 16):

    # Ask the user to enter the category of the current email
    category = input("Enter category for email " + str(i) + ": ")

    # Add the entered category to the list
    emails.append(category)
```

```python
# Assume the first category is the most frequently occurring category
dominant_category = emails[0]

# Count how many times the first category appears in the list
maximum_count = emails.count(dominant_category)
```

```python
# Go through each email category in the list
for category in emails:

    # Count how many times the current category appears in the list
    count = emails.count(category)

    # Check if the current category occurs more frequently
    # than the maximum count found so far
    if count > maximum_count:

        # Update the maximum count
        maximum_count = count

        # Update the dominant email category
        dominant_category = category
```

```python
# Calculate the percentage of emails belonging to the dominant category
percentage = (maximum_count / len(emails)) * 100

# Estimate how many emails out of the next 100 may belong to the dominant category
expected_emails = (percentage / 100) * 100
```

```python
# Display the most frequently occurring email category
print("\nDominant Email Category:", dominant_category)

# Display the number of emails in the dominant category
print("Number of Emails:", maximum_count)

# Display the percentage of emails in the dominant category
print("Percentage:", percentage, "%")

# Display the estimated number of dominant-category emails among the next 100 emails
print("Expected in Next 100 Emails:", round(expected_emails))
```
