**Task:**

Write a Python program that

* Accepts the performance rating of 11 employees on a scale of 1 to 5.
* Determines the most frequently assigned performance rating.
* Displays the most common performance rating.
* Assigns a salary increment percentage based on the most common rating:

      Rating 5 → 15% increment
      Rating 4 → 10% increment
      Rating 3 → 7% increment
      Rating 2 → 4% increment
      Rating 1 → 0% increment
* Accepts the current salary of an employee.
* Calculates the revised salary using the increment associated with the most common performance rating.
* Displays the revised salary.

```python
# Create an empty list to store the performance ratings of 11 employees
ratings = []
```

```python
4# Display a message asking the user to enter employee ratings
print("Enter performance ratings of 11 employees")

# Display the valid rating range
print("Rating should be between 1 and 5")

# Repeat the loop for 11 employees
for i in range(1, 12):

    # Ask the user to enter the rating of the current employee
    rating = int(input("Enter rating of employee " + str(i) + ": "))

    # Add the entered rating to the list
    ratings.append(rating)
```

```python
# Assume the first rating is the most frequently occurring rating
mode_rating = ratings[0]

# Count how many times the first rating appears in the list
maximum_count = ratings.count(mode_rating)
```

```python
# Go through each rating in the list
for rating in ratings:

    # Count how many times the current rating appears in the list
    count = ratings.count(rating)

    # Check if the current rating occurs more frequently
    # than the maximum count found so far
    if count > maximum_count:

        # Update the maximum count
        maximum_count = count

        # Update the most common performance rating
        mode_rating = rating
```

```python
# Display all the performance ratings entered
print("\nPerformance Ratings:", ratings)

# Display the most frequently occurring performance rating
print("Most Common Performance Rating:", mode_rating)

# Display the number of employees with the most common rating
print("Number of Employees with this Rating:", maximum_count)
```

```python
# Determine the salary increment based on the most common rating
if mode_rating == 5:

    # Give a 15% increment for a rating of 5
    increment = 15

elif mode_rating == 4:

    # Give a 10% increment for a rating of 4
    increment = 10

elif mode_rating == 3:

    # Give a 7% increment for a rating of 3
    increment = 7

elif mode_rating == 2:

    # Give a 4% increment for a rating of 2
    increment = 4

else:

    # Give no increment for a rating of 1
    increment = 0
```

```python
# Display the salary increment percentage
print("Salary Increment:", increment, "%")
```

```python
# Ask the user to enter the employee's current salary
salary = float(input("\nEnter current salary of the employee: ₹"))

# Calculate the amount of salary increment
# Formula: Increment Amount = Salary × Increment Percentage ÷ 100
increment_amount = salary * increment / 100

# Calculate the revised salary after adding the increment
# Formula: Revised Salary = Current Salary + Increment Amount
revised_salary = salary + increment_amount

# Display the increment amount, rounded to 2 decimal places
print("Increment Amount: ₹", round(increment_amount, 2))

# Display the revised salary, rounded to 2 decimal places
print("Revised Salary: ₹", round(revised_salary, 2))
```
