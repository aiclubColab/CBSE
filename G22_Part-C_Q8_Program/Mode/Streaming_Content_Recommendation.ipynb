**Task:**

Write a Python program for a streaming service that:

* Accepts the genre selected by 15 users from categories such as Action, Comedy, Drama, Thriller, or Animation.
* Uses the mode to identify the genre selected most frequently.
* Displays the most popular genre.
* Calculates the percentage of users who selected this genre.
* Based on this percentage, estimates how many users out of the next 100 users may select this genre.
* Displays a recommendation to the streaming service to promote more content from the popular genre.

```python
# Create an empty list to store the genres selected by 15 users
genres = []
```

```python
# Display a message asking the user to enter the selected genres
print("Enter the genre selected by 15 users:")

# Display the available genre options
print("Options: Action, Comedy, Drama, Thriller, Animation")

# Repeat the loop for 15 users
for i in range(1, 16):

    # Ask the user to enter the genre selected by the current user
    genre = input("Enter genre for user " + str(i) + ": ")

    # Add the entered genre to the list
    genres.append(genre)
```

```python
# Find the mode, which is the most frequently occurring genre
# Assume the first genre is the most popular genre
popular_genre = genres[0]

# Count how many times the first genre appears in the list
maximum_count = genres.count(popular_genre)
```

```python
# Go through each genre in the list
for genre in genres:

    # Count how many times the current genre appears in the list
    count = genres.count(genre)

    # Check if the current genre occurs more frequently
    # than the maximum count found so far
    if count > maximum_count:

        # Update the maximum count
        maximum_count = count

        # Update the most popular genre
        popular_genre = genre
```

```python
# Calculate the percentage of users who selected the most popular genre
percentage = (maximum_count / len(genres)) * 100

# Estimate the number of users who may select this genre
# among the next 100 users
estimated_users = (percentage / 100) * 100
```

```python
# Display the most popular genre
print("\nMost Popular Genre:", popular_genre)

# Display the number of users who selected the most popular genre
print("Number of Users:", maximum_count)

# Display the percentage of users who selected the most popular genre
print("Percentage of Users:", percentage, "%")

# Display the estimated number of users who may select this genre
# among the next 100 users
print("Estimated Users in Next 100:", round(estimated_users))

# Display a heading for the recommendation
print("\nRecommendation:")

# Suggest promoting more content of the most popular genre
print("Promote more", popular_genre, "content on the streaming platform.")
```
