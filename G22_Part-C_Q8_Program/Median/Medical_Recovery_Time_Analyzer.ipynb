**Task:**

Write a Python program that

* Accepts the recovery time in days for 7 patients who underwent a similar treatment.
* Arranges the recovery times in ascending order.
* Determines the median recovery time.
* Accepts the recovery time of a new patient.
* Compares the new patient's recovery time with the median recovery time.
* Displays "Within Typical Recovery Time" if the new patient's recovery time is equal to or less than the median; otherwise displays "Longer Recovery Expected".

```python
# Create an empty list to store the recovery times of 7 patients
recovery_times = []
```

```python
# Display a message asking the user to enter recovery times
print("Enter recovery time for 7 patients:")

# Repeat the loop for 7 patients
for i in range(1, 8):

    # Ask the user to enter the recovery time in days for the current patient
    days = float(input("Enter recovery time for patient " + str(i) + " in days: "))

    # Add the recovery time to the list
    recovery_times.append(days)
```

```python
# Arrange the recovery times from lowest to highest
recovery_times.sort()

# Find the middle value of the sorted list as the median recovery time
median_time = recovery_times[len(recovery_times) // 2]
```

```python
# Display the recovery times in ascending order
print("\nRecovery Times in Ascending Order:", recovery_times)

# Display the median recovery time in days
print("Median Recovery Time:", median_time, "days")
```

```python
# Ask the user to enter the recovery time of a new patient in days
new_patient = float(input("\nEnter recovery time of the new patient in days: "))
```

```python
# Compare the new patient's recovery time with the median recovery time
if new_patient <= median_time:

    # Display this message if the new patient's recovery time is within the typical time
    print("Within Typical Recovery Time")

# If the new patient's recovery time is greater than the median
else:

    # Display this message if a longer recovery is expected
    print("Longer Recovery Expected")
```
