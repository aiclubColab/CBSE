**Task:**

Write a Python program that

* Accepts the mileage of 7 vehicles in kilometres per litre (km/L).
* Arranges the mileage values in ascending order.
* Determines the median mileage.
* Accepts the distance of a planned journey in kilometres.
* Estimates the amount of fuel required for the journey using the median mileage.
* Accepts the price of fuel per litre.
* Calculates the estimated fuel cost for the journey.

```python
# Create an empty list to store the mileage of 7 vehicles
mileage = []
```

```python
# Display a message asking the user to enter mileage values
print("Enter the mileage of 7 vehicles in km/L:")

# Repeat the loop for 7 vehicles
for i in range(1, 8):

    # Ask the user to enter the mileage of the current vehicle
    value = float(input("Enter mileage of vehicle " + str(i) + ": "))

    # Add the entered mileage to the list
    mileage.append(value)
```

```python
# Arrange the mileage values from lowest to highest
mileage.sort()

# Find the middle value of the sorted list as the median mileage
median_mileage = mileage[len(mileage) // 2]
```

```python
# Display the mileage values in ascending order
print("\nMileage in Ascending Order:", mileage)

# Display the median mileage in kilometres per litre
print("Median Mileage:", median_mileage, "km/L")
```

```python
# Ask the user to enter the journey distance in kilometres
distance = float(input("\nEnter the journey distance in km: "))

# Ask the user to enter the fuel price for one litre
fuel_price = float(input("Enter fuel price per litre: ₹"))

# Calculate the amount of fuel required for the journey
# Formula: Fuel Required = Distance ÷ Mileage
fuel_required = distance / median_mileage

# Calculate the estimated fuel cost
# Formula: Fuel Cost = Fuel Required × Fuel Price
fuel_cost = fuel_required * fuel_price
```

```python
# Display the estimated fuel required, rounded to 2 decimal places
print("\nEstimated Fuel Required:", round(fuel_required, 2), "litres")

# Display the estimated fuel cost, rounded to 2 decimal places
print("Estimated Fuel Cost: ₹", round(fuel_cost, 2))
```
