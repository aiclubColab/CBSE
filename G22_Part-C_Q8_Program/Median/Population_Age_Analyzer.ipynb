**Task:**

Write a Python program that

* Accepts the ages of 9 people from a locality.
* Arranges the ages in ascending order.
* Determines the median age of the group.
* Accepts the age of a new person.
* Compares the new person's age with the median age.
* Displays whether the new person is younger than, equal to, or older than the median age.

```python
# Create an empty list to store the ages of 9 people
ages = []
```

```python
# Display a message asking the user to enter the ages
print("Enter the ages of 9 people:")

# Repeat the loop for 9 people
for i in range(1, 10):

    # Ask the user to enter the age of the current person
    age = int(input("Enter age of person " + str(i) + ": "))

    # Add the entered age to the list
    ages.append(age)
```

```python
# Arrange the ages from lowest to highest
ages.sort()

# Find the middle value of the sorted list as the median age
median_age = ages[len(ages) // 2]
```

```python
# Display the ages in ascending order
print("\nAges in Ascending Order:", ages)

# Display the median age
print("Median Age:", median_age)
```

```python
# Ask the user to enter the age of a new person
new_age = int(input("\nEnter the age of a new person: "))
```

```python
# Compare the new person's age with the median age
if new_age < median_age:

    # Display this message if the new person is younger than the median
    print("The new person is younger than the median age.")

# Check if the new person's age is greater than the median
elif new_age > median_age:

    # Display this message if the new person is older than the median
    print("The new person is older than the median age.")

# If the new person's age is equal to the median
else:

    # Display this message if both ages are equal
    print("The new person's age is equal to the median age.")
```
