**Task:**

Write a Python program that

* Accepts the baggage processing time in minutes for 9 passengers.
* Arranges the processing times in ascending order.
* Determines the median baggage processing time.
* Accepts the number of passengers expected for an upcoming flight.
* Estimates the total baggage processing time using the median processing time.
* Converts the estimated processing time from minutes to hours.
* Displays the median processing time and estimated total processing time.

```python
# Processing time is the time taken to process a passenger's baggage
# Processing time is measured in minutes

# Create an empty list to store the baggage processing times of 9 passengers
processing_times = []
```

```python
# Display a message asking the user to enter processing times
print("Enter baggage processing time for 9 passengers:")

# Repeat the loop for 9 passengers
for i in range(1, 10):

    # Ask the user to enter the processing time in minutes for the current passenger
    time = float(input("Enter processing time for passenger " + str(i) + " in minutes: "))

    # Add the entered processing time to the list
    processing_times.append(time)
```

```python
# Arrange the processing times from lowest to highest
processing_times.sort()

# Find the middle value of the sorted list as the median processing time
median_time = processing_times[len(processing_times) // 2]
```

```python
# Display the processing times in ascending order
print("\nProcessing Times in Ascending Order:", processing_times)

# Display the median processing time in minutes
print("Median Processing Time:", median_time, "minutes")
```

```python
# Ask the user to enter the number of passengers for the upcoming flight
passengers = int(input("\nEnter the number of passengers for the upcoming flight: "))
```

```python
# Estimate the total baggage processing time in minutes
# Formula: Total Time = Median Processing Time × Number of Passengers
total_minutes = median_time * passengers

# Convert the total processing time from minutes to hours
# Formula: Total Hours = Total Minutes ÷ 60
total_hours = total_minutes / 60
```

```python
# Display the estimated total processing time in minutes
print("Estimated Total Processing Time:", total_minutes, "minutes")

# Display the estimated total processing time in hours,
# rounded to 2 decimal places
print("Estimated Total Processing Time:", round(total_hours, 2), "hours")
```
