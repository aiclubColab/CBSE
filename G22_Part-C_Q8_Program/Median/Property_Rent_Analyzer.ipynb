**Task:**

Write a Python program that

* Accepts the monthly rent of 7 properties in a locality.
* Arranges the rents in ascending order.
* Determines the median monthly rent.
* Accepts the user's maximum rental budget.
* Compares the budget with the median rent.
* Displays "Budget Suitable" if the budget is equal to or greater than the median rent; otherwise displays "Look for a Lower-Rent Area".

```python
# Create an empty list to store the monthly rents of 7 properties
rents = []
```

```python
# Display a message asking the user to enter monthly rents
print("Enter the monthly rent of 7 properties:")

# Repeat the loop for 7 properties
for i in range(1, 8):

    # Ask the user to enter the monthly rent for the current property
    rent = float(input("Enter rent for property " + str(i) + ": "))

    # Add the entered rent to the list
    rents.append(rent)
```

```python
# Arrange the rents from lowest to highest
rents.sort()

# Find the middle value of the sorted list as the median rent
median_rent = rents[len(rents) // 2]
```

```python
# Display the rents in ascending order
print("\nRents in ascending order:", rents)

# Display the median monthly rent in Indian Rupees
print("Median Monthly Rent: ₹", median_rent)
```

```python
# Ask the user to enter their maximum rental budget
budget = float(input("\nEnter your maximum rental budget: ₹"))
```

```python
# Compare the user's budget with the median rent
if budget >= median_rent:

    # Display this message if the budget is equal to or higher than the median rent
    print("Budget Suitable")

# If the budget is lower than the median rent
else:

    # Suggest looking for an area with lower rent
    print("Look for a Lower-Rent Area")
```
